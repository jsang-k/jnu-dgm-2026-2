```python
#@title ▶ 준비 / Setup. 먼저 실행하세요 / Run this first { display-mode: "form" }
import math, os, gzip, struct, urllib.request, time, json, hashlib, platform, datetime
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
try:
    get_ipython().run_line_magic('matplotlib', 'inline')
except NameError:
    pass
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

torch.set_num_threads(2)                 # Colab CPU 기준. 같은 기기에서 다시 돌리면 같은 숫자 / the usual Colab CPU; a re-run on the same machine gives the same numbers
np.set_printoptions(precision=3, suppress=True)

NOTEBOOK = "A2 GAN track v13 (2026-10-17)"
SEED = 0                                 # 6주차 실습 §4 와 같은 시드 / the Week 6 lab §4 seed
N_TRAIN = 20000                      # 학습 이미지 수. 60,000장 가운데 앞 20,000장 / training images: the first 20,000 of the 60,000
EPOCHS = 3                        # 에폭 수 / epochs
BATCH = 128                              # 한 걸음에 쓰는 이미지 수. 6주차 실습 §4 와 같다 / images per step, as in the Week 6 lab §4
Z_DIM = 16                               # 잡음 z 의 차원 / dimension of the noise z
D_LR = 2e-4                              # D 의 learning rate. 네 학습에서 같다 / D's learning rate, the same in all four runs
A_LR, B_LR, C_NG = 2e-4, 1e-3, 3   # A 와 C 의 G learning rate, B 의 G learning rate, C 의 G 걸음 수 / G's learning rate in A and C, in B, and C's G steps per D step
Q4_OPTIONS = {"lr 4e-5": (4e-5, 1, "lr 4e-5, 1 step"), "lr 4e-3": (4e-3, 1, "lr 4e-3, 1 step"), "lr 3e-2": (3e-2, 1, "lr 3e-2, 1 step"), "2 G steps": (2e-4, 2, "lr 2e-4, 2 steps")}                  # Q4 에서 고를 수 있는 설정 / the settings Q4 offers
PRESCRIBED = {"SEED": 0, "N_TRAIN": 20000, "EPOCHS": 3, "BATCH": 128, "Z_DIM": 16, "D_LR": 2e-4, "A_LR": 2e-4, "B_LR": 1e-3, "C_NG": 3}
RUN_NAMES = {"A": "run A · G lr 2e-4, 1 G step (baseline)", "B": "run B · G lr 1e-3, 1 G step", "C": "run C · G lr 2e-4, 3 G steps"}
HEAD = {"A": "A · lr 2e-4, 1 step", "B": "B · lr 1e-3, 1 step", "C": "C · lr 2e-4, 3 steps"}   # 표와 그림의 머리 / column heads
COLORS = {"A": "#406551", "B": "#C9971C", "C": "#B34F3F", "Q4": "#4A6FA5"}   # 강의 슬라이드의 초록 · 금색 · 빨강, Q4 는 파랑 / the slides' green, gold and red; blue for Q4
RESULTS, RUNS = {}, {}                   # 이번 실행의 결과와 학습 기록 / this run's results and training records
Z64 = torch.randn(64, Z_DIM, generator=torch.Generator().manual_seed(7))   # 그림마다 같은 z 64개. 6주차 실습 §4 와 같다 / the same 64 z for every picture, as in the Week 6 lab §4
OUT_DIR = "a2_outputs"                   # 그림과 그림 PDF 를 저장하는 폴더 / where the figures and the figure PDF are saved
os.makedirs(OUT_DIR, exist_ok=True)


class Gen(nn.Module):
    """6주차 실습 §4 의 생성자 그대로. z(16) -> 128 x 7 x 7 -> 64 x 14 x 14 -> 1 x 28 x 28, 출력은 tanh 라 −1 ~ 1.
    The Week 6 lab §4 generator, unchanged: z(16) -> 128 x 7 x 7 -> 64 x 14 x 14 -> 1 x 28 x 28; the tanh output is in −1..1."""
    def __init__(self, zd=Z_DIM):
        super().__init__()
        self.fc = nn.Sequential(nn.Linear(zd, 128 * 7 * 7), nn.BatchNorm1d(128 * 7 * 7), nn.ReLU())
        self.up = nn.Sequential(nn.ConvTranspose2d(128, 64, 4, 2, 1), nn.BatchNorm2d(64), nn.ReLU(),   # 7 -> 14
                                nn.ConvTranspose2d(64, 1, 4, 2, 1), nn.Tanh())                          # 14 -> 28

    def forward(self, z):
        return self.up(self.fc(z).view(-1, 128, 7, 7))


def make_D():
    """6주차 실습 §4 의 판별자 그대로. 보폭 2 합성곱 둘로 28 -> 14 -> 7, 그다음 logit 하나 / the Week 6 lab §4 discriminator: two stride-2 convolutions, then one logit"""
    return nn.Sequential(nn.Conv2d(1, 64, 4, 2, 1), nn.LeakyReLU(0.2),                           # 28 -> 14
                         nn.Conv2d(64, 128, 4, 2, 1), nn.BatchNorm2d(128), nn.LeakyReLU(0.2),    # 14 -> 7
                         nn.Flatten(), nn.Linear(128 * 7 * 7, 1))


def bce(logits, target):
    """이진 교차엔트로피. logit에서 바로 계산한다. bce(logit, 1) = 평균 −log σ(a), bce(logit, 0) = 평균 −log(1 − σ(a)). 6주차 실습과 같다.
    Binary cross-entropy computed straight from the logits: bce(logit, 1) = mean −log σ(a), bce(logit, 0) = mean −log(1 − σ(a)). As in the Week 6 lab."""
    return F.binary_cross_entropy_with_logits(logits, torch.full_like(logits, float(target)))


def _fn(name):
    """학생 함수가 통과했거나 기준 코드로 바뀌었으면 그것을, 아니면 노트북의 기준 함수를 쓴다 / the student's function when usable, else the notebook's reference"""
    return globals()[name] if code_usable(CODE_OF[name]) else REFS[name]


def train_gan(g_lr, n_g, name, student):
    """6주차 실습 §4 의 학습 반복문. 한 걸음 = D 한 번 갱신(코드 1), 그다음 같은 x_fake 로 G 를 n_g 번 갱신(코드 2).
    The Week 6 lab §4 training loop: one step = one D update (Code 1), then n_g G updates starting from the same x_fake (Code 2).
    이미지 순서와 D 의 걸음에 쓰는 잡음은 따로 고정한 난수열 rng 에서 뽑는다. 그래서 C 의 추가 G 걸음이 새 잡음을 뽑아도 A · B · C 의 이미지 순서와 그 잡음은 같다.
    The image order and the noise of each D step come from their own fixed random stream rng, so C's extra G steps, which draw new noise, leave them the same in A, B and C.
    에폭이 끝날 때마다 고정 z 64개로 이미지를 찍어 둔다(코드 3). 학습 A 는 노트북의 기준 함수로, B · C · Q4 는 여러분의 함수로 돈다.
    At the end of every epoch the 64 fixed z are decoded (Code 3). Run A uses the notebook's reference functions; B, C and Q4 use yours.
    돌려주는 것: 에폭별 이미지, 에폭별 (D 손실, G 손실, D(x) 평균, D(G(z)) 평균). Returns the images per epoch and, per epoch, (D loss, G loss, mean D(x), mean D(G(z)))."""
    d_step_fn = _fn("d_step") if student else REFS["d_step"]
    g_steps_fn = _fn("g_steps") if student else REFS["g_steps"]
    gen_fn = _fn("generate_images") if student else REFS["generate_images"]
    torch.manual_seed(SEED)                  # 같은 시드 = 같은 처음 가중치. C 의 추가 걸음은 이 난수열에서 새 잡음을 뽑는다 / same seed: same initial weights; C's extra steps draw their new noise from this stream
    G, D = Gen(), make_D()
    rng = torch.Generator().manual_seed(SEED)   # 이미지 순서와 D 걸음의 잡음만 뽑는 난수열 / a stream used only for the image order and the noise of each D step
    opt_G = torch.optim.Adam(G.parameters(), lr=g_lr, betas=(0.5, 0.999))   # G 만 움직인다. betas 는 DCGAN 논문의 값 / moves only G; the DCGAN paper's betas
    opt_D = torch.optim.Adam(D.parameters(), lr=D_LR, betas=(0.5, 0.999))   # D 만 움직인다 / moves only D
    snaps, hist = {}, []
    print(f"--- {name}")
    print(f'{"epoch":>5} {"D loss":>8} {"G loss":>8} {"D(x)":>7} {"D(G(z))":>8} {"time":>7}')
    t0 = time.time()
    for ep in range(1, EPOCHS + 1):
        perm = torch.randperm(N_TRAIN, generator=rng)
        d_tot = g_tot = dr_tot = df_tot = 0.0; n = 0
        for i in range(0, N_TRAIN, BATCH):
            x_real = X_TRAIN[perm[i:i + BATCH]]          # 진짜 이미지 한 배치 / a batch of real images
            x_fake = G(torch.randn(len(x_real), Z_DIM, generator=rng))  # 생성한 이미지. G 까지 이어진 계산 그래프를 단다 / generated images; they carry the graph back to G
            d_loss = d_step_fn(D, opt_D, x_real, x_fake)   # 코드 1. D 가 갱신되기 직전의 손실 / Code 1: the loss just before D's update
            with torch.no_grad():                          # 갱신된 D 의 판정, 같은 배치. 코드 2 의 G 가 받는 것 / the updated D's verdicts on the same batch, what G faces in Code 2
                dr_tot += torch.sigmoid(D(x_real)).mean().item(); df_tot += torch.sigmoid(D(x_fake)).mean().item()
            g_loss = g_steps_fn(G, D, opt_G, x_fake, n_g)  # 코드 2. G 의 첫 갱신에서 잰 손실 / Code 2: the loss of G's first update
            d_tot += float(d_loss); g_tot += float(g_loss); n += 1
        snaps[ep] = np.asarray(gen_fn(G, Z64), dtype=np.float32)   # 코드 3 / Code 3
        hist.append((ep, d_tot / n, g_tot / n, dr_tot / n, df_tot / n))
        print(f"{ep:5d} {d_tot / n:8.3f} {g_tot / n:8.3f} {dr_tot / n:7.3f} {df_tot / n:8.3f} {time.time() - t0:6.0f}s")
    return snaps, hist


def settings_now():
    """지금 쓰는 설정값 / the settings in use now"""
    return {k: globals()[k] for k in PRESCRIBED}


def run_condition(key, g_lr, n_g, name, student):
    """학습 하나를 돌리고 RUNS 에 남긴다. 오류가 나도 모두 실행을 멈추지 않는다.
    Run one training and keep it in RUNS. An error does not stop Run all."""
    t0 = time.time()
    try:
        snaps, hist = train_gan(g_lr, n_g, name, student)
    except Exception as e:  # noqa: BLE001
        print(f"학습이 끝나지 못했습니다 / the run did not finish: {type(e).__name__}: {e}")
        RUNS.pop(key, None)
        return
    sec = time.time() - t0
    RUNS[key] = {"snaps": snaps, "hist": hist, "sec": sec, "g_lr": g_lr, "n_g": n_g, "name": name, "student": student,
                 "id": f"{key}-{time.time():.3f}", "settings": settings_now(), "code": {c: code_state(c) for c in CODE_CELLS},
                 "hash": img_hash(snaps[EPOCHS]), "plan": plan_now(), "xplan": xplan_now()}   # 학습할 때의 예측을 남긴다 / keeps the plan as it was at training time
    print(f"걸린 시간 / time: {sec:.0f} s")
    final = snaps[EPOCHS]
    if not np.isfinite(final).all():
        print("생성한 이미지에 nan 이나 inf 가 있습니다. 학습이 무너졌습니다. 코드 1 · 2 를 다시 보세요.")
        print("The generated images contain nan or inf, so training broke down. Look at Code 1 and 2 again.")
    elif final.min() < -1e-6 or final.max() > 1 + 1e-6:
        print("이미지 값이 0 ~ 1 을 벗어납니다. 코드 3 의 범위 변환을 다시 보세요. / Pixel values leave 0..1: check the range conversion in Code 3.")


def img_hash(imgs):
    """64장의 지문. 그림이 바뀌면 바뀐다 / a fingerprint of the 64 images; it changes when the pictures change"""
    return hashlib.sha256(np.round(np.asarray(imgs, dtype=np.float32), 4).tobytes()).hexdigest()[:8]


def plan_now():
    """지금 기록된 진단 · 예측 / the diagnosis and predictions recorded now"""
    return {k: RESULTS.get("Plan", {}).get(k) for k in PLAN_FIELDS}


def xplan_now():
    """지금 기록된 Q4 설정 · 예측 / the Q4 setting and prediction recorded now"""
    return {k: RESULTS.get("ExtraPlan", {}).get(k) for k in ("extra", "predict", "look")}


def spread(imgs):
    """64장 사이의 평균 픽셀 차이: 이미지 두 장씩 모든 쌍에서 픽셀 값 차이의 절댓값을 평균한 것. 모든 z 가 같은 그림이면 0 에 가깝다. 표준 지표는 아니다.
    The mean absolute pixel difference over all pairs of the 64 images: near 0 when every z gives the same picture. Not a standard metric."""
    a = torch.as_tensor(np.asarray(imgs), dtype=torch.float32).reshape(len(imgs), -1)
    return float(torch.cdist(a, a, p=1).sum() / (len(a) * (len(a) - 1) * a.shape[1]))


def grey_share(imgs):
    """회색 픽셀 비율: 값이 0.2 ~ 0.8 인 픽셀의 비율. 덱 「Did We Get Sharper Samples?」 의 검사. 표준 지표는 아니다.
    Grey-pixel share: the share of pixels with values between 0.2 and 0.8, the check on the slide "Did We Get Sharper Samples?". Not a standard metric."""
    a = np.asarray(imgs); return float(((a > 0.2) & (a < 0.8)).mean())


def knn_digits(imgs, k=5):
    """이미지마다 학습 이미지 20,000장 가운데 가장 가까운 k 장의 숫자 다수결. 덱 노트의 최근접 이웃 검사. 잡음도 어느 숫자로든 읽는다.
    For each image, the majority digit of its k nearest images among the first 20,000 training images, the deck notes' nearest-neighbour check. It reads noise as some digit too."""
    a = torch.as_tensor(np.asarray(imgs), dtype=torch.float32).reshape(len(imgs), -1)
    idx = torch.cdist(a, KNN_BANK).topk(k, largest=False).indices
    pred = torch.mode(KNN_LABELS[idx], dim=1).values.numpy()
    return np.bincount(pred, minlength=10).tolist()


def draw_grid(ax, imgs, n, title, cols=8, ids=True):
    """n 장을 cols 열 격자로. 번호 01 부터 / n images in a grid of cols columns, numbered from 01"""
    rows = math.ceil(n / cols)
    canvas = np.ones((rows * 30, cols * 30), dtype=np.float32)
    for k in range(n):
        r, c = divmod(k, cols)
        canvas[r * 30 + 1:r * 30 + 29, c * 30 + 1:c * 30 + 29] = np.asarray(imgs[k]).reshape(28, 28)
    ax.imshow(canvas, cmap="gray", vmin=0, vmax=1, interpolation="nearest")
    if ids:
        for k in range(n):
            r, c = divmod(k, cols)
            ax.text(c * 30 + 2, r * 30 + 8, f"{k + 1:02d}", color="#FFD54A", fontsize=8, fontweight="bold")
    ax.set_title(title, fontsize=10); ax.axis("off")


print("torch", torch.__version__, "| numpy", np.__version__)
print("실행 장치 / device: cpu. GPU 가 없어도 됩니다 / no GPU needed")
```

```python
#@title ▶ 코드 칸 점검 / Checks for the code cells. 고치지 마세요 / Do not edit { display-mode: "form" }
# 코드 1 ~ 3 을 노트북의 기준 함수와 한 걸음씩 맞춰 보는 함수들입니다. 여러분이 고칠 곳은 없습니다.
# Functions that check Code 1 to 3 against the notebook's reference functions, one step at a time. There is nothing here for you to edit.
import copy, unicodedata

CODE_CELLS = ["Code1", "Code2", "Code3"]
CODE_OF = {"d_step": "Code1", "g_steps": "Code2", "generate_images": "Code3"}
OPENS = {c: ("학습 B · C · Q4. 세 학습은 코드 칸 셋이 모두 끝나야 열립니다", "runs B, C and Q4, which open only when all three code cells are done") for c in ("Code1", "Code2", "Code3")}
PLAN_FIELDS = ["STUDENT_ID", "DIAGNOSIS", "PREDICT_B", "PREDICT_C", "DECISION_RULE"]
EXTRA_FIELDS = ["EXTRA_PREDICT", "EXTRA_LOOK"]
BLANK = ("아직 None 인 빈칸이 있습니다. None 을 코드로 바꾸세요.", "A blank (None) is still there. Replace None with your code.")


def _blank_msg(name):
    return (f"빈칸 {name} 이 아직 None 입니다. None 을 코드로 바꾸고 이 셀을 다시 실행하세요.", f"The blank {name} is still None. Replace None with your code and run this cell again.")
LINE = "-" * 96


def record(name, **values):
    """이번 실행의 결과를 남긴다 / keep a result of this run"""
    RESULTS[name] = values


def code_ok(name):
    """그 코드 칸이 여러분의 코드로 통과했나 / did that code cell pass with your own code?"""
    return RESULTS.get(name, {}).get("ok") is True


def code_usable(name):
    """여러분의 코드가 통과했거나, 기준 코드를 쓰기로 했나 / did your code pass, or did you switch to the reference code?"""
    return code_ok(name) or RESULTS.get(name, {}).get("reference") is True


def code_state(name):
    return "OK" if code_ok(name) else ("reference" if code_usable(name) else "not done")


def _call(fname, *args):
    """학생 함수를 부른다. Python 오류가 나도 멈추지 않고 그 내용을 돌려준다.
    Call a student's function. On a Python error, return the message instead of stopping."""
    fn = globals().get(fname)
    if not callable(fn):
        return None, (f"함수 {fname} 가 없습니다. 함수 이름을 바꾸지 마세요.", f"There is no function {fname}. Do not rename it.")
    try:
        return fn(*args), None
    except NotImplementedError as e:          # 빈칸 표시 줄이 멈춘 것 / stopped by the line that marks a blank
        return None, _blank_msg(e)
    except Exception as e:  # noqa: BLE001
        msg = f"{type(e).__name__}: {e}"
        return None, (f"Python 오류가 났습니다. {msg}", f"Python raised an error. {msg}")


def _pad(s, width):
    """한글처럼 두 칸을 차지하는 글자를 세어 왼쪽 정렬로 채운다 / left-align, counting wide characters such as Hangul as two columns"""
    w = sum(2 if unicodedata.east_asian_width(ch) in "WF" else 1 for ch in s)
    return s + " " * max(0, width - w)


def _report(name, passed, why, info=()):
    """코드 칸 결과를 찍고 RESULTS 에 남긴다 / print a code cell's result and record it"""
    num = name[-1]
    ko, en = OPENS[name]
    print(LINE)
    if passed:
        print(f" 코드 {num} / Code {num}   통과 / passed")
        for s in info:
            print("   " + s)
    else:
        print(f" 코드 {num} / Code {num}   아직 통과하지 않았습니다 / not passed yet")
        for k, e in why:
            print("   " + k)
            if e != k:
                print("   " + e)
        print(f" 이 칸이 통과하면 열리는 것: {ko}. receipt가 READY 가 되려면 이 칸도 끝나 있어야 합니다. 끝내 통과하지 않으면 check_code{num}(use_reference=True) 로 바꿔도 되고, 감점은 없습니다. 아래 셀은 그대로 돌아갑니다.")
        print(f" What opens when this cell passes: {en}. The receipt's READY needs this cell done as well. If it never passes, switching with check_code{num}(use_reference=True) also works, at no cost. The cells below still run.")
    print(LINE)
    record(name, ok=bool(passed), reference=False)


def _use_reference(name, fname):
    """기준 코드로 바꾼다. 감점은 없고 receipt에 적힌다 / switch to the reference code; no penalty, and the receipt records it"""
    globals()[fname] = REFS[fname]
    num = name[-1]
    print(LINE)
    print(f" 코드 {num} / Code {num}   기준 코드를 씁니다 / using the reference code")
    print("   use_reference=True 이므로 노트북의 기준 코드로 바꿨습니다. 아래 학습과 결과는 그대로 이어집니다.")
    print("   use_reference=True, so the notebook's reference code is used. The runs and results below carry on.")
    print("   감점은 없습니다. receipt의 코드 칸 줄에 reference 로 적히고, PDF 에는 이 칸이 무엇을 계산하는지 똑같이 설명합니다.")
    print("   There is no penalty. The receipt's code-cell line records it as reference, and your PDF still explains what this cell computes.")
    print(LINE)
    record(name, ok=False, reference=True)


def _safe(check):
    """점검이 예상 못 한 값에 걸려도 실행이 멈추지 않게 한다 / keep a check from stopping the run on an unexpected value"""
    def run(*args, **kwargs):
        try:
            check(*args, **kwargs)
        except Exception as e:  # noqa: BLE001
            msg = f"{type(e).__name__}: {e}"
            _report("Code" + check.__name__[-1], False,
                    [(f"결과를 점검하지 못했습니다. 함수가 돌려주는 값의 모양을 확인하세요. {msg}",
                      f"The result could not be checked. Check the form of what your function returns. {msg}")])
    return run


# ---- 기준 함수 셋. 학습 A 가 쓰고, 기준 코드로 바꾼 칸에도 쓴다 / the three reference functions: run A uses them, and so does a cell switched to the reference ----
def _ref_d_step(D, opt_D, x_real, x_fake):
    logit_real = D(x_real)
    logit_fake = D(x_fake.detach())
    d_loss = bce(logit_real, 1) + bce(logit_fake, 0)
    opt_D.zero_grad(); d_loss.backward(); opt_D.step()
    return d_loss.item()


def _ref_g_steps(G, D, opt_G, x_fake, n_g):
    g_loss = bce(D(x_fake), 1)
    opt_G.zero_grad(); g_loss.backward(); opt_G.step()
    first_loss = g_loss.item()
    for _ in range(n_g - 1):
        x_fake = G(torch.randn(len(x_fake), Z_DIM))
        g_loss = bce(D(x_fake), 1)
        opt_G.zero_grad(); g_loss.backward(); opt_G.step()
    return first_loss


def _ref_generate_images(G, z):
    was_training = G.training
    G.eval()
    with torch.no_grad():
        imgs = (G(z) + 1) / 2
    G.train(was_training)
    return imgs.squeeze(1).cpu().numpy()


REFS = {"d_step": _ref_d_step, "g_steps": _ref_g_steps, "generate_images": _ref_generate_images}


def _fresh_pair(seed):
    """점검용 작은 세상: 시드를 고정해 G · D 와 작은 배치를 만든다 / a small fixed world for the checks: G, D and a small batch from a fixed seed"""
    torch.manual_seed(seed)
    G, D = Gen(), make_D()
    x_real = X_TRAIN[:6]
    return G, D, x_real


def _params_equal(m1, m2, tol=1e-5):
    return all(torch.allclose(a, b, atol=tol, rtol=1e-4) for a, b in zip(m1.parameters(), m2.parameters()))


def _params_moved(before, m):
    return any(not torch.equal(a, b) for a, b in zip(before, m.parameters()))


@_safe
def check_code1(use_reference=False):
    if use_reference:
        return _use_reference("Code1", "d_step")
    why, info = [], []
    G, D1, x_real = _fresh_pair(31)
    D0, D2 = copy.deepcopy(D1), copy.deepcopy(D1)       # D0: 갱신 전 사본, 오답 판별용 / a copy before the update, for diagnosing wrong answers
    o1 = torch.optim.SGD(D1.parameters(), lr=1e-3); o2 = torch.optim.SGD(D2.parameters(), lr=1e-3)
    with torch.no_grad():
        x_fake_src = G(torch.randn(6, Z_DIM))
    x_fake1 = x_fake_src.clone().requires_grad_(True)   # 기준 / reference
    x_fake2 = x_fake_src.clone().requires_grad_(True)   # 여러분 / yours
    want = _ref_d_step(D1, o1, x_real, x_fake1)
    out, err = _call("d_step", D2, o2, x_real, x_fake2)
    if err:
        why.append(err)
    elif out is None:
        why.append(BLANK)
    else:
        got = float(out) if not hasattr(out, "item") else float(out.item())
        if x_fake2.grad is not None:
            why.append(("D 의 손실이 x_fake 를 거슬러 G 쪽으로 기울기를 보냈습니다. 생성 이미지에 .detach() 를 붙여야 D 의 갱신이 G 를 건드리지 않습니다.",
                        "D's loss sent gradient back through x_fake toward G. Attach .detach() to the generated images so that D's update does not touch G."))
        elif not abs(got - want) < 1e-5:
            with torch.no_grad():
                swapped = (bce(D0(x_real), 0) + bce(D0(x_fake1.detach()), 1)).item()
            if abs(got - swapped) < 1e-4:
                why.append(("정답이 뒤집혔습니다. 진짜 이미지에 0, 생성 이미지에 1 을 주면 D 는 거꾸로 배웁니다.",
                            "The targets are swapped: with 0 for real and 1 for generated images, D learns the wrong way round."))
            elif abs(got - want / 2) < 1e-4:
                why.append(("두 항의 평균을 냈습니다. 덱 「D's Loss Is Binary Cross-Entropy」 처럼 두 항을 더합니다.",
                            "The two terms were averaged. Add them, as on the slide \"D's Loss Is Binary Cross-Entropy\"."))
            else:
                why.append((f"손실이 {got:.4f} 인데 기준은 {want:.4f} 입니다. 두 항(진짜에 1, 생성에 0)을 더한 것인지 확인하세요.",
                            f"The loss is {got:.4f} but the reference gives {want:.4f}. Check that both terms (1 for real, 0 for generated) are added."))
        elif not _params_equal(D1, D2):
            why.append(("손실 값은 맞지만 D 의 가중치가 기준과 다르게 움직였습니다. D(x_fake).detach() 처럼 logit을 떼어 내면 D 도 생성 이미지에서 배우지 못합니다. 떼어 낼 것은 x_fake 입니다. opt_D.zero_grad() · backward() · step() 세 줄도 확인하세요.",
                        "The loss value is right but D's weights moved differently from the reference. Detaching the logit, as in D(x_fake).detach(), stops D from learning on the generated images too; detach x_fake instead. Also check the three lines opt_D.zero_grad(), backward() and step()."))
    if not why:
        G, D1, x_real = _fresh_pair(32)
        D2 = copy.deepcopy(D1)
        o1 = torch.optim.SGD(D1.parameters(), lr=1e-3); o2 = torch.optim.SGD(D2.parameters(), lr=1e-3)
        xf = G(torch.randn(6, Z_DIM))
        want = _ref_d_step(D1, o1, x_real, xf.detach().clone().requires_grad_(True))
        xf2 = xf.detach().clone().requires_grad_(True)
        got, err = _call("d_step", D2, o2, x_real, xf2)
        if err or got is None or not abs(float(got) - want) < 1e-5 or not _params_equal(D1, D2):
            why.append(("두 번째 작은 배치에서 기준과 다릅니다.", "Differs from the reference on a second small batch."))
        else:
            info = [f"작은 배치 둘에서 손실 {want:.4f} 과 D 의 갱신이 기준과 같고, 생성 이미지 쪽으로는 기울기가 가지 않습니다 / on two small batches the loss {want:.4f} and D's update match the reference, and no gradient reaches the generated images"]
    _report("Code1", not why, why, info)


@_safe
def check_code2(use_reference=False):
    if use_reference:
        return _use_reference("Code2", "g_steps")
    why, info = [], []
    for n_g, seed, b in ((1, 41, 6), (3, 42, 5)):              # 배치 크기를 바꿔 len(x_fake) 를 쓰는지도 본다 / a different batch size also checks that len(x_fake) is used
        G1, D, _ = _fresh_pair(seed)
        G2 = copy.deepcopy(G1)
        o1 = torch.optim.SGD(G1.parameters(), lr=1e-3); o2 = torch.optim.SGD(G2.parameters(), lr=1e-3)
        d_before = [p.clone() for p in D.parameters()]
        torch.manual_seed(seed + 100); xf1 = G1(torch.randn(b, Z_DIM)); want = _ref_g_steps(G1, D, o1, xf1, n_g)
        torch.manual_seed(seed + 100); xf2 = G2(torch.randn(b, Z_DIM)); out, err = _call("g_steps", G2, D, o2, xf2, n_g)
        if err:
            if "second time" in err[1]:
                why.append(("같은 x_fake 를 다시 썼습니다. 첫 걸음의 backward 가 그 계산 그래프를 이미 지나갔으므로 다시 쓸 수 없습니다. 새 잡음을 G 에 넣어 새 이미지를 만드세요.",
                            "The same x_fake was reused. The first step's backward already went through its graph, so it cannot be used again. Make new images by feeding new noise to G."))
            elif "conv2d" in err[1] and ("3D" in err[1] or "4D" in err[1]):
                why.append(("잡음을 그대로 D 에 넣었습니다. D 는 28 × 28 이미지를 받습니다. 잡음 z 를 먼저 G 에 넣어 이미지 G(z) 로 만드세요.",
                            "Noise went straight into D, but D takes 28 × 28 images. Feed the noise z to G first, to get images G(z)."))
            else:
                why.append(err)
            break
        if out is None:
            why.append(BLANK); break
        got = float(out) if not hasattr(out, "item") else float(out.item())
        if _params_moved(d_before, D):
            why.append(("D 의 가중치가 움직였습니다. 코드 2 는 G 만 갱신합니다. opt_D 를 부르거나 D 의 매개변수를 바꾸지 마세요.",
                        "D's weights moved. Code 2 updates only G; do not call opt_D or change D's parameters.")); break
        if n_g == 1 and not abs(got - want) < 1e-5:
            alt0 = bce(D(xf2.detach()), 0).item(); altmm = (-bce(D(xf2.detach()), 0)).item()
            if abs(got - alt0) < 1e-4:
                why.append(("정답에 0 을 줬습니다. 그러면 G 는 D 를 돕는 쪽으로 움직입니다. G 는 D 가 생성 이미지를 진짜(1)라고 말하기를 원합니다.",
                            "The target is 0, so G would move to help D. G wants D to call the generated images real (1)."))
            elif abs(got - altmm) < 1e-4:
                why.append(("minimax 꼴 log(1 − D) 을 그대로 줄이는 식입니다. 실습과 덱 「What Loss Do We Actually Use for G?」 는 비포화 꼴 bce(D(x_fake), 1) 을 씁니다.",
                            "This is the minimax form, minimizing log(1 − D) as written. The lab and the slide \"What Loss Do We Actually Use for G?\" use the non-saturating form bce(D(x_fake), 1)."))
            else:
                why.append((f"n_g = 1 에서 손실이 {got:.4f} 인데 기준은 {want:.4f} 입니다. 첫 걸음은 주어진 x_fake 로 bce(D(x_fake), 1) 입니다.",
                            f"With n_g = 1 the loss is {got:.4f} but the reference gives {want:.4f}. The first step is bce(D(x_fake), 1) on the given x_fake."))
            break
        if not _params_equal(G1, G2):
            if n_g == 1:
                why.append(("손실 값은 맞지만 G 의 가중치가 움직이지 않았거나 다르게 움직였습니다. x_fake 에 .detach() 를 붙이면 기울기가 G 에 닿지 않습니다. opt_G.zero_grad() · backward() · step() 세 줄도 확인하세요.",
                            "The loss value is right but G's weights did not move, or moved differently. A .detach() on x_fake keeps the gradient from reaching G. Also check the three lines opt_G.zero_grad(), backward() and step()."))
            else:
                why.append(("n_g = 1 은 맞지만 n_g = 3 에서 G 가 기준과 다르게 움직였습니다. 두 번째 걸음부터는 새 잡음 z 를 이번 배치와 같은 개수(len(x_fake))만큼 뽑아 G 에 넣어 새 이미지를 만듭니다. 개수를 숫자로 적거나 같은 x_fake 를 다시 쓰면 안 됩니다.",
                            "n_g = 1 is right, but with n_g = 3 G moved differently from the reference. From the second step on, draw new noise z for as many images as this batch has (len(x_fake)) and feed it to G. Do not write the count as a number, and do not reuse the same x_fake."))
            break
    if not why:
        info = ["n_g = 1(배치 6)과 n_g = 3(배치 5)에서 G 의 갱신이 기준과 같고 D 는 움직이지 않습니다 / with n_g = 1 (batch 6) and n_g = 3 (batch 5), G's updates match the reference and D does not move"]
    _report("Code2", not why, why, info)


@_safe
def check_code3(use_reference=False):
    if use_reference:
        return _use_reference("Code3", "generate_images")
    why, info = [], []
    G, _, _ = _fresh_pair(51)
    G.train()
    z = torch.randn(5, Z_DIM, generator=torch.Generator().manual_seed(52))
    want = _ref_generate_images(G, z)
    before = [p.clone() for p in G.parameters()]
    out, err = _call("generate_images", G, z)
    if err:
        why.append(err)
    elif out is None:
        why.append(BLANK)
    else:
        got = np.asarray(torch.as_tensor(out).detach().cpu().numpy() if hasattr(out, "detach") else out, dtype=np.float32)
        mode_after = G.training                              # 여러분의 함수가 G 를 어느 모드로 남겼나 / the mode your function left G in
        G.train()
        with torch.no_grad():
            train_mode = ((G(z) + 1) / 2).squeeze(1).numpy()   # 학습 모드 그대로 디코딩한 것 / decoded without switching to eval mode
        if got.squeeze().shape != want.shape:
            why.append((f"크기가 {tuple(got.shape)} 입니다. (5, 28, 28), 곧 이미지마다 28 × 28 이어야 합니다. .squeeze(1) 로 채널 축을 없애세요.",
                        f"The shape is {tuple(got.shape)}; it must be (5, 28, 28), one 28 × 28 image each. Remove the channel axis with .squeeze(1)."))
        elif got.min() < -1e-6 or got.max() > 1 + 1e-6:
            why.append((f"값이 {got.min():.2f} ~ {got.max():.2f} 입니다. G 의 tanh 출력 −1 ~ 1 을 0 ~ 1 로 옮기세요: (G(z) + 1) / 2.",
                        f"The values run from {got.min():.2f} to {got.max():.2f}. Map G's tanh output from −1..1 to 0..1: (G(z) + 1) / 2."))
        elif not mode_after:
            why.append(("G 가 평가 모드로 남았습니다. 그림을 그린 뒤 원래 모드로 되돌려야 다음 학습 걸음이 배치 정규화를 제대로 씁니다.",
                        "G was left in evaluation mode. Restore the original mode after drawing, so the next training step uses batch normalization properly."))
        elif np.allclose(got.squeeze(), train_mode, atol=1e-5) and not np.allclose(got.squeeze(), want, atol=1e-5):
            why.append(("학습 모드로 디코딩했습니다. 그림을 그릴 때는 G.eval() 로 바꿔야 배치 정규화가 배치 통계 대신 저장된 평균을 씁니다(6주차 실습 §4).",
                        "Decoded in training mode. Switch to G.eval() when drawing, so batch normalization uses its stored averages instead of the batch statistics (Week 6 lab §4)."))
        elif not np.allclose(got.squeeze(), want, atol=1e-5):
            why.append((f"기준과 최대 {np.abs(got.squeeze() - want).max():.3g} 만큼 다릅니다. eval 모드, no_grad, (G(z) + 1) / 2 를 확인하세요.",
                        f"Differs from the reference by up to {np.abs(got.squeeze() - want).max():.3g}. Check eval mode, no_grad and (G(z) + 1) / 2."))
        elif _params_moved(before, G):
            why.append(("G 의 가중치가 바뀌었습니다. 그림을 그릴 때는 학습하지 않습니다.", "G's weights changed. Drawing must not train."))
    if not why:
        info = ["평가 모드에서 디코딩한 기준 이미지와 같고, 값은 0 ~ 1, G 는 학습 모드로 돌아왔습니다 / matches the reference decoded in eval mode, values in 0..1, G back in training mode"]
    _report("Code3", not why, why, info)


def plan_ready():
    """학번 · 진단 · 예측 둘 · decision rule이 모두 채워졌나 / are the student ID, the diagnosis, the two predictions and the decision rule all filled in?"""
    return all(isinstance(globals().get(k), str) and globals()[k].strip() for k in PLAN_FIELDS)


def show_plan():
    """진단과 예측을 찍고 남긴다 / print and keep the diagnosis and predictions"""
    texts = {k: (globals().get(k) or "").strip() if isinstance(globals().get(k), str) else "" for k in PLAN_FIELDS}
    for k, v in texts.items():
        print(f"{k:14s} {v if v else '<비어 있음 / still empty>'}")
    print(f"{'WROTE_BEFORE_RUNNING':20s} {bool(globals().get('WROTE_BEFORE_RUNNING'))}")
    if "A" not in RUNS:
        print("\n학습 A 를 먼저 실행하고 그 그림을 본 뒤에 쓰세요. / Run A first and look at its pictures before writing.")
    elif plan_ready():
        print("\n기록했습니다. §2 의 코드 1 ~ 3 이 통과하거나 기준 코드로 바뀌면 학습 B · C 가 열립니다. 결과를 본 뒤에는 예측을 고치지 마세요. 고치면 receipt에 표시됩니다.")
        print("Recorded. Runs B and C open once Code 1 to 3 in §2 pass or are switched to the reference code. Do not edit your predictions after you see the results; the receipt shows it if you do.")
    else:
        print("\n다섯 칸을 모두 채우고 이 셀을 다시 실행하세요. 그래야 B · C 가 열립니다. 예측이 틀려도 괜찮습니다.")
        print("Fill in all five fields and run this cell again; B and C open after that. Wrong predictions are fine.")
    record("Plan", **texts, before=bool(globals().get("WROTE_BEFORE_RUNNING")), after_A=("A" in RUNS))


def not_open(need_codes=True):
    """학습이 아직 열리지 않았으면 이유를 찍고 True / if a run is not open yet, print why and return True"""
    if "A" not in RUNS:
        print("아직 열리지 않았습니다. §1 의 학습 A 를 먼저 실행하세요. / Not open yet. Run A in §1 first.")
        return True
    if plan_ready() and RESULTS.get("Plan") is not None and not RESULTS["Plan"].get("after_A"):
        print("아직 열리지 않았습니다. Q1 · Q2 폼을 학습 A 보다 먼저 실행했습니다. A 의 그림을 본 뒤 폼 셀을 다시 실행하세요.")
        print("Not open yet. The Q1 · Q2 form was run before run A. Look at A's pictures, then run the form cell again.")
        return True
    if not plan_ready() or RESULTS.get("Plan") is None:
        print("아직 열리지 않았습니다. Q1 · Q2 폼의 다섯 칸을 채우고 그 셀을 실행한 뒤 이 셀을 다시 실행하세요.")
        print("Not open yet. Fill in the five fields of the Q1 · Q2 form, run that cell, then run this cell again.")
        return True
    if need_codes and not all(code_usable(c) for c in CODE_CELLS):
        left = [c[-1] for c in CODE_CELLS if not code_usable(c)]
        print(f"아직 열리지 않았습니다. 코드 {', '.join(left)} 이 통과하거나 기준 코드로 바뀌면 이 학습이 열립니다.")
        print(f"Not open yet. This run opens after Code {', '.join(left)} passes or is switched to the reference code.")
        return True
    return False


def numbers_table(keys):
    """조건마다 마지막 에폭의 학습 숫자와 64장의 두 검사 / per run, the last epoch's training numbers and the two checks on the 64 images"""
    rows = {}
    for k in keys:
        ep, dl, gl, dr, df = RUNS[k]["hist"][-1]
        imgs = RUNS[k]["snaps"][EPOCHS]
        rows[k] = {"d_loss": round(dl, 3), "g_loss": round(gl, 3), "D_real": round(dr, 3), "D_fake": round(df, 3),
                   "grey": round(grey_share(imgs), 4), "spread": round(spread(imgs), 3), "knn": knn_digits(imgs), "hash": RUNS[k]["hash"]}
    return rows


def parse_labels(text):
    """쉼표나 빈칸으로 나눈 판정 열여섯 개를 읽는다. 잘못되면 None 과 이유 / read sixteen labels separated by commas or spaces; None and the reason when they are wrong"""
    if not isinstance(text, str) or not text.strip():
        return None, ("비어 있음", "empty")
    labels = [t.strip().lower() for t in text.replace(";", ",").replace(",", " ").split() if t.strip()]
    if len(labels) != 16:
        return None, (f"{len(labels)}개입니다. 열여섯 개여야 합니다", f"{len(labels)} labels; sixteen are needed")
    bad = [t for t in labels if t not in list("0123456789") + ["x"]]
    if bad:
        return None, (f"{bad[0]!r} 은 0 ~ 9 나 x 가 아닙니다", f"{bad[0]!r} is not 0 to 9 or x")
    return labels, None


def audit_summary(labels):
    readable = [t for t in labels if t != "x"]
    return {"labels": labels, "readable": len(readable), "kinds": len(set(readable))}


print("점검 함수 준비 / checks ready: check_code1, check_code2, check_code3")
```

이 손실은 G 를 움직이면 안 됩니다. x_fake 는 G 까지 이어진 계산 그래프를 달고 있어서, 그대로 D 에 넣으면 D 의 손실이 G 쪽으로도 기울기를 보냅니다. 실습 §2 에서 무엇을 붙였는지, 덱 「Four Ways to Get This Wrong」 의 1번과 「Checkpoint. Who Sees What?」 의 둘째 질문을 떠올리세요.
*This loss must not move G. x_fake carries the computation graph back to G, so feeding it to D as it is would send D's gradient toward G as well. Recall what you attached in the lab §2, item 1 of the slide "Four Ways to Get This Wrong", and the second question of "Checkpoint. Who Sees What?".*


```python
# ---------------------------------------------------------------- 코드 1 / Code 1 ----
# D 의 한 걸음 / One D step  (6주차 덱 · 실습 §2 / Week 6 deck and lab §2)
#
# 다음 이름을 씁니다 / Names used below
#   D           : 판별자. D(x) 는 이미지마다 logit 하나, 크기 (B, 1) / the discriminator; D(x) gives one logit per image, shape (B, 1)
#   opt_D       : D 의 가중치만 움직이는 Adam / the Adam optimizer that moves only D's weights
#   x_real      : 진짜 이미지 한 배치. 크기 (B, 1, 28, 28) / a batch of real images, shape (B, 1, 28, 28)
#   x_fake      : G 가 만든 이미지. G 까지 이어진 계산 그래프를 달고 있다 / images G made; they carry the graph back to G
#   logit_real  : 진짜 이미지에 낸 logit (주어짐) / the logits on the real images (given)
#   logit_fake  : 생성 이미지에 낸 logit. G 쪽으로는 기울기가 가면 안 된다   <- 빈칸 / the logits on the generated images; no gradient may flow toward G   <- blank
#   d_loss      : 두 BCE 항을 더한 D 의 손실, 숫자 하나   <- 빈칸 / D's loss, the two BCE terms added, one number   <- blank
#
# 참고 / Reference
#   bce(logits, 1) = 평균 −log σ(a) / the batch mean of −log σ(a);   bce(logits, 0) = 평균 −log(1 − σ(a)) / the batch mean of −log(1 − σ(a))
#     예 / e.g.  bce(torch.tensor([[0.0]]), 1)  ->  tensor(0.6931)   (= ln 2: logit 0 은 확률 ½ / a logit of 0 is a probability of ½)
#   t.detach() : t 와 같은 값을 갖되 기울기 계산에서는 상수로 취급되는 텐서. 그 앞으로는 기울기가 가지 않는다
#                a tensor with the same values as t, treated as a constant for gradients; no gradient flows back past it
#     예 / e.g.  a = torch.tensor(3., requires_grad=True);  (2 * a.detach() + a).backward();  a.grad  ->  tensor(1.)

def d_step(D, opt_D, x_real, x_fake):
    """D 를 한 번 갱신하고 손실을 돌려준다 / update D once and return the loss"""
    logit_real = D(x_real)                     # 진짜 이미지의 logit / logits on the real images
    # TODO. 두 줄의 None 을 코드로 바꾸세요. 생성 이미지의 logit(G 를 건드리지 않게), 그리고 두 항을 더한 손실.
    # TODO. Replace None in the two lines: the logits on the generated images (without touching G), then the loss that adds the two terms.
    logit_fake = None
    if logit_fake is None: raise NotImplementedError("logit_fake")   # 빈칸이 남아 있으면 여기서 멈춘다. 지우지 마세요 / stops here while the blank is empty; do not delete
    d_loss = None
    if d_loss is None: raise NotImplementedError("d_loss")           # 위와 같다 / as above

    opt_D.zero_grad()                          # 지난 걸음의 기울기를 지운다 / clear last step's gradients
    d_loss.backward()                          # D 의 가중치에 기울기를 쌓는다 / accumulate gradients on D's weights
    opt_D.step()                               # D 만 한 걸음 움직인다 / move only D, one step
    return d_loss.item()                       # 숫자로 돌려준다 / return it as a plain number


check_code1()
# 기대값 / Expected   작은 배치 둘에서 손실과 D 의 갱신이 기준과 같고, 생성 이미지 쪽으로 기울기가 가지 않으면 통과입니다.
#   It passes when the loss and D's update match the reference on two small batches and no gradient reaches the generated images.
#   「G 쪽으로 기울기를 보냈다」 가 나오면 생성 이미지에 .detach() 가 없는 것입니다.
#   "sent gradient toward G" means the generated images lack .detach().
```

### 코드 2 / Code 2. G 의 걸음 n_g 번 / n_g G steps

덱 「What Loss Do We Actually Use for G?」 의 비포화 손실로 G 를 갱신합니다. 첫 걸음은 코드 1 과 같은 x_fake 를 씁니다. 이번에는 기울기가 D 를 지나 G 까지 가야 하므로 .detach() 를 붙이지 않습니다. 실습 §2 의 두 번째 줄입니다.
*Update G with the non-saturating loss of the slide "What Loss Do We Actually Use for G?". The first step uses the same x_fake as Code 1. This time the gradient must pass through D into G, so no .detach(). It is the second line of the lab §2.*

n_g 가 1 보다 크면 그 뒤로 n_g − 1 번 더 갱신합니다. 걸음마다 새 잡음 z 로 새 이미지를 만듭니다. 조건 C 와 Q4 가 이 횟수를 바꿉니다. D 는 이 함수 안에서 움직이지 않습니다. 함수가 돌려주는 손실은 첫 갱신의 것입니다. 방금 D 가 판정한 같은 배치에서 잰 값이라, 걸음 수가 다른 학습끼리도 같은 뜻으로 견줄 수 있습니다.
*If n_g is larger than 1, update n_g − 1 more times, each on new images from new noise z. Condition C and Q4 change this count. D does not move inside this function. The loss the function returns is that of the first update: it is measured on the same batch D has just judged, so runs with different step counts report the same thing.*


```python
# ---------------------------------------------------------------- 코드 2 / Code 2 ----
# G 의 걸음 n_g 번 / n_g G steps  (6주차 덱 · 실습 §2 / Week 6 deck and lab §2)
#
# 다음 이름을 씁니다 / Names used below
#   G, D     : 생성자, 판별자 / the generator and the discriminator
#   opt_G    : G 의 가중치만 움직이는 Adam / the Adam optimizer that moves only G's weights
#   x_fake   : 코드 1 과 같은 생성 이미지. 계산 그래프가 G 까지 이어져 있다 / the same generated images as in Code 1; the graph reaches G
#   n_g      : 이 걸음에서 G 를 갱신하는 횟수. A · B 는 1, C 는 3 / how many times G is updated in this step: 1 in A and B, 3 in C
#   g_loss   : 비포화 G 손실, 숫자 하나   <- 빈칸 / the non-saturating G loss, one number   <- blank
#   first_loss : 첫 갱신의 손실. 이 함수가 돌려주는 값 (주어짐) / the loss of the first update, which the function returns (given)
#   G(z)     : 잡음 z 로 G 가 만든 이미지, 크기 (n, 1, 28, 28) / the images G makes from noise z, shape (n, 1, 28, 28)
#   Z_DIM    : 잡음 z 의 차원, 16 / the dimension of z, 16
#
# 참고 / Reference
#   bce(logits, target), 코드 1 과 같다 / as in Code 1.   예 / e.g.  bce(torch.tensor([[2.0]]), 0)  ->  tensor(2.1269)  (= ln(1 + e²))
#   torch.randn(n, Z_DIM) : 표준정규분포에서 뽑은 잡음 n 개, 크기 (n, Z_DIM) / n noise vectors from the standard normal, shape (n, Z_DIM)
#   len(x_fake) : 배치 크기 B / the batch size B

def g_steps(G, D, opt_G, x_fake, n_g):
    """같은 x_fake 로 G 를 한 번, 그 뒤 새 잡음으로 n_g − 1 번 더 갱신한다. 첫 갱신의 손실을 돌려준다.
    Update G once on the same x_fake, then n_g − 1 more times on new noise. Return the loss of the first update."""
    # TODO. None 을 코드로 바꾸세요. G 는 D 가 생성 이미지를 무엇이라고 말하기를 원하나요? 그 정답으로 BCE 한 항입니다.
    # TODO. Replace None. What does G want D to say about the generated images? One BCE term with that target.
    g_loss = None
    if g_loss is None: raise NotImplementedError("g_loss")   # 빈칸이 남아 있으면 여기서 멈춘다. 지우지 마세요 / stops here while the blank is empty; do not delete

    opt_G.zero_grad(); g_loss.backward(); opt_G.step()   # G 만 한 걸음 / one step of G only
    first_loss = g_loss.item()                 # 이 걸음의 G 손실로 남기는 값 / the value kept as this step's G loss

    for _ in range(n_g - 1):                   # 나머지 걸음. n_g = 1 이면 돌지 않는다 / the remaining steps; none when n_g = 1
        # TODO. None 을 코드로 바꾸세요. 새 잡음으로 새 이미지를 만듭니다. 같은 x_fake 를 다시 쓰지 않습니다.
        # TODO. Replace None: make new images from new noise. Do not reuse the same x_fake.
        x_fake = None
        if x_fake is None: raise NotImplementedError("x_fake")   # 위와 같다 / as above
        g_loss = bce(D(x_fake), 1)             # 위와 같은 손실 / the same loss as above
        opt_G.zero_grad(); g_loss.backward(); opt_G.step()
    return first_loss


check_code2()
# 기대값 / Expected   n_g = 1 과 n_g = 3 에서 G 의 갱신이 기준과 같고 D 가 움직이지 않으면 통과입니다.
#   It passes when G's updates match the reference with n_g = 1 and n_g = 3, and D does not move.
#   「정답에 0 을 줬다」 가 나오면 G 가 D 를 돕는 쪽으로 움직이는 손실입니다.
#   "The target is 0" means a loss that moves G to help D.
```

### 코드 3 / Code 3. 고정 z 로 이미지 그리기 / Draw images from the fixed z

6주차 실습 §4 의 그림 셀이 한 일입니다. G 에는 배치 정규화가 있습니다. 실습처럼 그림을 그릴 때는 평가 모드로 바꿉니다. 그러면 배치 정규화가 그 배치의 통계 대신 학습 중 저장한 평균을 써서, 그림 한 장이 다른 z 와 상관없이 자기 z 로만 정해지고, 그리는 동안 저장한 평균도 바뀌지 않습니다. 그리고 tanh 출력 −1 ~ 1 을 픽셀 값 0 ~ 1 로 옮깁니다. 그린 뒤에는 원래 모드로 되돌려 학습이 이어지게 합니다.

```python
# ---------------------------------------------------------------- 코드 3 / Code 3 ----
# 고정 z 로 이미지 그리기 / Draw images from the fixed z  (6주차 실습 §4 / Week 6 lab §4)
#
# 다음 이름을 씁니다 / Names used below
#   G             : 생성자. 배치 정규화가 있다 / the generator, with batch normalization
#   z             : 고정 잡음. 크기 (n, 16) / the fixed noise, shape (n, 16)
#   was_training  : 들어올 때의 모드. True 면 학습 모드 (주어짐) / the mode on entry; True means training mode (given)
#   imgs          : 이미지 값 0 ~ 1, 크기 (n, 1, 28, 28)   <- 빈칸 / pixel values in 0..1, shape (n, 1, 28, 28)   <- blank
#
# 참고 / Reference
#   model.eval()  : 평가 모드로. 배치 정규화가 저장된 평균을 쓴다 / evaluation mode: batch normalization uses its stored averages
#   model.train(flag) : flag 가 True 면 학습 모드, False 면 평가 모드 / training mode when flag is True, evaluation mode when False
#   with torch.no_grad(): ... : 그 안에서는 기울기를 기록하지 않는다 / inside the block no gradients are recorded
#   G(z) 의 값은 −1 ~ 1 (tanh) / G(z) is in −1..1 (tanh).   예 / e.g.  (torch.tensor([-1.0, 0.0, 1.0]) + 1) / 2  ->  tensor([0.0, 0.5, 1.0])

def generate_images(G, z):
    """고정 z 를 디코딩해 0 ~ 1 값의 이미지를 돌려준다 / decode the fixed z and return images with values in 0..1"""
    was_training = G.training                  # 들어올 때의 모드를 기억한다 / remember the mode on entry
    # TODO. 첫 None 줄은 G 를 평가 모드로 바꾸는 한 줄, 둘째 None 은 tanh 출력을 0 ~ 1 로 옮긴 이미지입니다.
    # TODO. The first None line switches G to evaluation mode; the second None is the images with the tanh output mapped to 0..1.
    None
    with torch.no_grad():
        imgs = None
        if imgs is None: raise NotImplementedError("imgs")   # 빈칸이 남아 있으면 여기서 멈춘다. 지우지 마세요 / stops here while the blank is empty; do not delete

    G.train(was_training)                      # 원래 모드로 되돌린다 / restore the original mode
    return imgs.squeeze(1).cpu().numpy()       # (n, 28, 28) 의 numpy 배열로 / as a numpy array of shape (n, 28, 28)


check_code3()
# 기대값 / Expected   평가 모드에서 디코딩한 기준 이미지와 같고 값이 0 ~ 1 이며 G 가 학습 모드로 돌아오면 통과입니다.
#   It passes when the images match the reference decoded in eval mode, values are in 0..1, and G is back in training mode.
#   「학습 모드로 디코딩했다」 가 나오면 G.eval() 이 빠진 것입니다.
#   "Decoded in training mode" means G.eval() is missing.
```

## 4 · 선택 자료 / Optional ring illustration

MNIST와 다른 dataset에서 mode coverage를 살펴보는 추가 예시입니다. 실행과 해석은 선택이며 Q3(c), 점수, READY에 필요하지 않습니다. 보고서에서는 MNIST 관찰을 우선하세요.

This supplementary example illustrates mode coverage on a different dataset. It is optional and not required for Q3(c), marks or READY. Base the assessed answer on MNIST. These ring runs do not establish that one G loss is generally better.

```python
#@title §4 ▶ 고리 자료 두 번 학습 / Two ring runs  (약 2분 / about two minutes) { display-mode: "form" }
RUN_RING = False #@param {type:"boolean"}
if RUN_RING:
    RING_R, RING_STD, RING_STEPS, RING_B, RING_RADIUS, RING_MIN = 2.0, 0.05, 10000, 256, 0.2, 50
    RING_SNAPS, RING_PROBE = (1000, 3000, 5000, 10000), (1, 2, 5, 10, 20, 50) + tuple(range(100, 10001, 100))
    CENTERS = torch.tensor([[RING_R * math.cos(2 * math.pi * k / 8), RING_R * math.sin(2 * math.pi * k / 8)] for k in range(8)])
    Z_RING = torch.randn(1000, Z_DIM, generator=torch.Generator().manual_seed(1234))


    def sample_ring(n):
        """진짜 점 n 개: 군집 하나를 고르고 그 중심 둘레에 작은 잡음 / n real points: pick a cluster, add small noise around its center"""
        return CENTERS[torch.randint(0, 8, (n,))] + RING_STD * torch.randn(n, 2)


    def ring_nets():
        """6주차 실습 §1 의 작은 G 와 D / the small G and D of the Week 6 lab §1"""
        G = nn.Sequential(nn.Linear(Z_DIM, 128), nn.ReLU(), nn.Linear(128, 128), nn.ReLU(), nn.Linear(128, 2))
        D = nn.Sequential(nn.Linear(2, 128), nn.ReLU(), nn.Linear(128, 128), nn.ReLU(), nn.Linear(128, 1))
        return G, D


    def cluster_counts(x):
        """군집마다, 가장 가까운 중심이 그 군집이고 거리가 0.2 보다 작은 점의 수 / per cluster, the points whose nearest center is that cluster and lie closer than 0.2"""
        d_min, nearest = torch.cdist(x, CENTERS).min(dim=1)
        return torch.bincount(nearest[d_min < RING_RADIUS], minlength=8)


    def train_ring(g_loss_fn, name):
        """6주차 실습 §1 의 학습. 한 걸음 = D 한 번(비포화 · minimax 모두 같은 D 손실), 그다음 같은 x_fake 로 G 한 번 / the Week 6 lab §1 training: one D step (the same D loss in both runs), then one G step on the same x_fake"""
        torch.manual_seed(2)                                       # 실습 §1 과 같은 시드 / the lab §1 seed
        G, D = ring_nets()
        opt_G = torch.optim.Adam(G.parameters(), lr=1e-3, betas=(0.5, 0.999)); opt_D = torch.optim.Adam(D.parameters(), lr=1e-3, betas=(0.5, 0.999))
        snaps, probes = {}, {}
        print(f"--- {name}")
        for step in range(1, RING_STEPS + 1):
            x_real, x_fake = sample_ring(RING_B), G(torch.randn(RING_B, Z_DIM))
            d_loss = bce(D(x_real), 1) + bce(D(x_fake.detach()), 0)
            opt_D.zero_grad(); d_loss.backward(); opt_D.step()
            if step in RING_PROBE:
                with torch.no_grad():
                    probes[step] = torch.sigmoid(D(x_fake)).mean().item()    # 갱신된 D 의 판정 평균 / the updated D's mean verdict on the generated points
            g_loss = g_loss_fn(D, x_fake)
            opt_G.zero_grad(); g_loss.backward(); opt_G.step()
            if step in RING_SNAPS:
                with torch.no_grad():
                    snaps[step] = G(Z_RING)
                c = cluster_counts(snaps[step])
                print(f"  step {step:6,d}   덮은 군집 / covered {int((c >= RING_MIN).sum())} of 8   군집 밖 / off-cluster {1 - c.sum().item() / 1000:.1%}")
        return {"snaps": snaps, "probes": probes}


    if set(globals().get("RING", {})) == {"ns", "mm"}:
        print("고리 학습은 이 런타임에서 이미 끝났습니다. 여러분의 코드와 상관없는 학습이라 다시 돌리지 않습니다.")
        print("The ring runs already finished in this runtime. They do not depend on your code, so they are not repeated.")
    else:
        RING = {}
        t0 = time.time()
        RING["ns"] = train_ring(lambda D, xf: bce(D(xf), 1), "ring run 1 · non-saturating G loss (the lab's)")
        RING["mm"] = train_ring(lambda D, xf: -bce(D(xf), 0), "ring run 2 · minimax G loss (the original paper's)")
        print(f"걸린 시간 / time: {time.time() - t0:.0f} s")
else:
    print("선택 ring을 건너뜁니다 / Optional ring skipped; no effect on marks or READY.")
```

```python
#@title 선택 탐색 도구 / Optional exploration tools { display-mode: "form" }
EXP_TRACK = 'GAN'
# Optional exploration tools. Core RUNS / RESULTS / HEAD are read-only here.
import uuid, csv, zipfile
from pathlib import Path
from IPython.display import display, FileLink

EXPLORATIONS = globals().get('EXPLORATIONS', [])
EXP_ROOT = Path(OUT_DIR) / 'explorations'
EXP_ROOT.mkdir(parents=True, exist_ok=True)
# Restore this runtime's disk records if the helper cell is run again.
_known_exp = {r['id'] for r in EXPLORATIONS}
for _path in sorted(EXP_ROOT.glob('*/record.json')):
    _saved_exp = json.loads(_path.read_text(encoding='utf-8'))
    if _saved_exp['id'] not in _known_exp:
        EXPLORATIONS.append(_saved_exp)
        _known_exp.add(_saved_exp['id'])

def exploration_core_rows():
    rows = []
    table = RESULTS.get('Table4', {}).get('rows', {})
    for key in ('A', 'B', 'C', 'Q4'):
        if key not in RUNS or key not in table:
            continue
        r = {'id': key, 'status': 'complete', **table[key]}
        if EXP_TRACK == 'VAE':
            r['beta'] = RUNS[key]['beta']
        else:
            r.update(g_lr=RUNS[key]['g_lr'], n_g=RUNS[key]['n_g'])
        rows.append(r)
    return rows

def exploration_tables():
    """Show controlled comparisons; never mix both GAN factors into one ranking."""
    from IPython.display import HTML
    import html
    rows = exploration_core_rows() + [r for r in EXPLORATIONS if r['status'] == 'complete']
    if EXP_TRACK == 'VAE':
        groups = [('β comparison', sorted(rows, key=lambda r: (r['beta'], r['id'])), ['beta','rec','kl','inside','spread'])]
    else:
        groups = [
            ('G learning rate · one G update per D update', sorted([r for r in rows if r['n_g']==1], key=lambda r:(r['g_lr'],r['id'])), ['g_lr','n_g','g_loss','D_fake','spread']),
            ('G update count · G learning rate 2e-4', sorted([r for r in rows if r['g_lr']==A_LR], key=lambda r:(r['n_g'],r['id'])), ['g_lr','n_g','g_loss','D_fake','spread'])]
    for title, group, fields in groups:
        cols = ['id'] + fields + ['kNN kinds']
        body = '<h4>'+html.escape(title)+'</h4><table><tr>'+''.join('<th>'+html.escape(k)+'</th>' for k in cols)+'</tr>'
        for row in group:
            vals = [row['id']] + [f'{row[k]:.4g}' if isinstance(row.get(k),(int,float)) else str(row.get(k,'')) for k in fields] + [str(sum(n>=3 for n in row.get('knn',[])))]
            body += '<tr>'+''.join('<td>'+html.escape(v)+'</td>' for v in vals)+'</tr>'
        display(HTML(body+'</table>'))
    print('kNN 종류 수는 readability 점수가 아닙니다. 그림으로 확인하세요. / kNN kinds are not readability scores; inspect the images.')
    failed = [r['id'] for r in EXPLORATIONS if r['status']=='failed']
    if failed: print('실패 기록도 보존 / Failed attempts retained:', ', '.join(failed))
    return rows

def export_explorations():
    rows = exploration_core_rows() + EXPLORATIONS
    (EXP_ROOT/'all_records.json').write_text(json.dumps(rows,ensure_ascii=False,indent=2),encoding='utf-8')
    fields = sorted(set(k for r in rows for k in r)) if rows else ['id']
    with (EXP_ROOT/'comparison.csv').open('w',encoding='utf-8-sig',newline='') as f:
        writer=csv.DictWriter(f,fieldnames=fields);writer.writeheader()
        for r in rows:writer.writerow({k:json.dumps(v,ensure_ascii=False) if isinstance(v,(list,dict)) else v for k,v in r.items()})
    archive = Path(OUT_DIR)/f'A2_{EXP_TRACK}_exploration_records.zip'
    with zipfile.ZipFile(archive,'w',zipfile.ZIP_DEFLATED) as z:
        for p in EXP_ROOT.rglob('*'):
            if p.is_file(): z.write(p,p.relative_to(EXP_ROOT))
        # Include baseline figures so the download supports comparison outside Colab.
        for p in Path(OUT_DIR).glob(f'{EXP_TRACK}_*.png'):z.write(p,'required_figures/'+p.name)
    return archive

def show_exploration_history(show_all=False):
    """모든 기록의 목록, 그리고 가장 최근 기록의 그림(show_all 이면 모든 기록의 그림)을 다시 보여 준다
    List every record, then re-display the figures of the latest record (of every record if show_all)."""
    from IPython.display import Image
    if not EXPLORATIONS:
        return
    print('탐색 기록 / Exploration records:')
    for entry in EXPLORATIONS:
        print(f"  요청 / request {entry.get('request', '-')} | {entry['id']} | {entry['option']} | {entry['status']}")
    if not show_all and len(EXPLORATIONS) > 1:
        print('그림은 가장 최근 기록만 보여 줍니다. 모든 기록의 그림은 SHOW_ALL_HISTORY 를 켜면 보입니다. / Figures of the latest record only; turn on SHOW_ALL_HISTORY to see every record.')
    for entry in (EXPLORATIONS if show_all else EXPLORATIONS[-1:]):
        print(entry['id'], '|', entry['option'], '|', entry['status'])
        print('예측 / Prediction:', entry['prediction'])
        if entry.get('error'): print(entry['error'])
        for name in entry.get('files',[]):
            if name.endswith('.png') and (Path(OUT_DIR)/name).exists():
                display(Image(filename=str(Path(OUT_DIR)/name)))

def run_exploration(option, prediction, request=None):
    if not isinstance(prediction,str) or not prediction.strip():
        print('실행 전 예측을 적으세요. 학습하지 않았습니다. / Enter a prediction before running; no training started.')
        return None
    if option not in Q4_OPTIONS:
        print('Q4와 같은 네 선택지 중 하나를 고르세요. / Choose one of the four Q4 options.')
        return None
    if any(k not in RUNS for k in ('A','B','C','Q4')) or not all(code_usable(c) for c in CODE_CELLS):
        print('필수 A·B·C·Q4와 코드 칸을 먼저 완료하세요. / Complete A, B, C, Q4 and the code cells first.')
        return None
    if settings_now()!=PRESCRIBED:
        print('필수 설정을 원래 값으로 돌려놓으세요. / Restore the prescribed settings first.')
        return None
    # 요청 번호 하나는 한 번만 학습한다: 모두 실행을 다시 해도 같은 탐색이 되풀이되지 않는다
    # One request number trains once, so Run all does not repeat an exploration
    try:
        req = int(request)
    except (TypeError, ValueError):
        req = 0
    if req < 1:
        print('EXP_REQUEST 에 1 이상의 요청 번호를 적으세요. 학습하지 않았습니다. / Set EXP_REQUEST to a request number of 1 or more; no training started.')
        return None
    done = next((r for r in EXPLORATIONS if r.get('request') == req), None)
    if done is not None:
        nxt = max(int(r.get('request') or 0) for r in EXPLORATIONS) + 1
        print(f"요청 번호 {req}: 이미 실행한 요청입니다 ({done['id']} · {done['option']} · {done['status']}). 같은 요청은 다시 학습하지 않습니다. 새로 실행하려면 EXP_REQUEST 를 {nxt} 처럼 아직 쓰지 않은 번호로 바꾸세요.")
        print(f"Request {req} already ran ({done['id']} · {done['option']} · {done['status']}); the same request is not trained again. To run again, set EXP_REQUEST to an unused number such as {nxt}.")
        return done
    run_id = 'EXP_' + datetime.datetime.now().strftime('%Y%m%d_%H%M%S') + '_' + uuid.uuid4().hex[:8]
    folder = EXP_ROOT/run_id
    folder.mkdir(exist_ok=False)
    entry = {'id':run_id,'request':req,'option':option,'prediction':prediction.strip(),'status':'running','settings':settings_now(), 'files':[]}
    t0=time.time()
    (folder/'record.json').write_text(json.dumps(entry,ensure_ascii=False,indent=2),encoding='utf-8')
    try:
        # Preserve the core run's global torch RNG state as well as its results.
        with torch.random.fork_rng(devices=[]):
            if EXP_TRACK=='VAE':
                beta,label=Q4_OPTIONS[option]
                model,hist=train_vae(beta,run_id,student=True)
                ev=evaluate(model)
                imgs=_fn('generate_from_prior')(model,64,Z_SEED)
                entry.update(beta=beta,rec=ev['rec'],kl=ev['kl'],inside=ev['inside'],max_mu=ev['max_mu'],kl_coord=[float(v) for v in _fn('kl_per_coordinate')(ev['mu'],ev['logvar'])])
                fig,axes=plt.subplots(2,8,figsize=(10,3))
                draw_rows(axes,[X_REAL8,reconstruct(model,X_REAL8,3)],['real',label])
                fig.tight_layout();fig.savefig(folder/'reconstruction.png',dpi=120,bbox_inches='tight');plt.close(fig)
                entry['files'].append(str((folder/'reconstruction.png').relative_to(Path(OUT_DIR))))
            else:
                g_lr,n_g,label=Q4_OPTIONS[option]
                snaps,hist=train_gan(g_lr,n_g,run_id,student=True)
                imgs=snaps[EPOCHS]
                _,dl,gl,dr,df=hist[-1]
                entry.update(g_lr=g_lr,n_g=n_g,d_loss=dl,g_loss=gl,D_real=dr,D_fake=df)
                fig,axes=plt.subplots(1,EPOCHS,figsize=(4*EPOCHS,4),squeeze=False)
                for ep in range(1,EPOCHS+1):draw_grid(axes[0,ep-1],snaps[ep][:16],16,f'{label} · epoch {ep}',cols=4)
                fig.tight_layout();fig.savefig(folder/'epochs.png',dpi=120,bbox_inches='tight');plt.close(fig)
                entry['files'].append(str((folder/'epochs.png').relative_to(Path(OUT_DIR))))
            entry.update(grey=grey_share(imgs),spread=spread(imgs),knn=knn_digits(imgs),hash=img_hash(imgs),history=hist)
            if not np.isfinite(np.asarray(imgs)).all() or not np.isfinite(np.asarray(hist)).all():raise ValueError('Non-finite training result')
            np.savez_compressed(folder/'samples.npz',samples=np.asarray(imgs))
            fig,ax=plt.subplots(figsize=(8,8));draw_grid(ax,imgs,64,f'{run_id}\n{label}')
            fig.tight_layout();fig.savefig(folder/'samples.png',dpi=120,bbox_inches='tight');plt.close(fig)
            entry['files'] += [str((folder/n).relative_to(Path(OUT_DIR))) for n in ['samples.png','samples.npz']]
        entry['status']='complete'
    except Exception as exc:
        entry.update(status='failed',error=f'{type(exc).__name__}: {exc}')
        print('탐색 실패를 기록했습니다. 필수 결과는 유지됩니다. / Exploration failure recorded; required results are unchanged.',entry['error'])
    entry['seconds']=round(time.time()-t0,2)
    (folder/'record.json').write_text(json.dumps(entry,ensure_ascii=False,indent=2),encoding='utf-8')
    EXPLORATIONS.append(entry)
    print('탐색 기록 저장 / Saved exploration:', run_id, '· 요청 / request', req)
    return entry
```
