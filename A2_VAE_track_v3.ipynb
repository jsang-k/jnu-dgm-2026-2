```python
#@title ▶ 준비 / Setup. 먼저 실행하세요 / Run this first { display-mode: "form" }
import math, os, gzip, struct, urllib.request, time, json, hashlib, platform, datetime
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.distributions import Normal, kl_divergence
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.lines import Line2D

torch.set_num_threads(2)                 # Colab CPU 기준. 같은 기기에서 다시 돌리면 같은 숫자 / the usual Colab CPU; a re-run on the same machine gives the same numbers
np.set_printoptions(precision=3, suppress=True)

NOTEBOOK = "A2 VAE track v3 (2026-10-15)"
SEED = 0                                 # 4주차 실습과 같은 시드 / the Week 4 lab seed
LATENT = 2                               # 잠재 차원. 모든 코드를 평면에 그릴 수 있다 / latent dimensions, so every code fits on a flat page
EPOCHS = 6                               # 4주차 실습 §4 와 같다 / as in §4 of the Week 4 lab
BATCH = 256                              # 한 걸음에 쓰는 이미지 수 / images per step
LR = 1e-3                                # Adam 학습률 / Adam learning rate
WARM = 2                                 # warm-up 이 β 를 0 에서 1 로 올리는 에폭 수 / epochs over which the warm-up raises β from 0 to 1
BETA_LOW = 0.1                           # 학습 3 의 β / β of run 3
R = 3.0                                  # 원의 반지름. 4주차 실습과 같다 / the circle's radius, as in the Week 4 lab
P_INSIDE = 1 - math.exp(-R ** 2 / 2)     # 2차원 N(0, I) 표본이 반지름 R 안에 들 확률 0.989 / probability that a 2-D N(0, I) draw lands inside radius R
PRESCRIBED = {"SEED": 0, "LATENT": 2, "EPOCHS": 6, "BATCH": 256, "LR": 1e-3, "WARM": 2, "BETA_LOW": 0.1, "R": 3.0}
RUN_NAMES = {"beta1": "run 1 · beta = 1 (baseline)", "warmup": "run 2 · KL warm-up", "beta01": "run 3 · beta = 0.1"}
COLORS = {"beta1": "#406551", "warmup": "#C9971C", "beta01": "#B34F3F"}   # 강의 슬라이드의 초록 · 금색 · 빨강 / the slides' green, gold and red
HEAD = {"beta1": "1 · beta = 1", "warmup": "2 · warm-up", "beta01": "3 · beta = 0.1"}   # 표와 그림의 머리 / column heads for tables and figures
RESULTS, RUNS = {}, {}                   # 이번 실행의 결과와 학습한 모형 / this run's results and the trained models


class VAE(nn.Module):
    """4주차 실습의 VAE 그대로 / the Week 4 lab VAE, unchanged.
    인코더 784 -> 512 -> 256 -> (mu, logvar), 디코더 잠재 -> 256 -> 512 -> 784 logit.
    Encoder 784 -> 512 -> 256 -> (mu, logvar); decoder latent -> 256 -> 512 -> 784 logits."""
    def __init__(self, latent):
        super().__init__()
        self.latent = latent
        self.enc = nn.Sequential(nn.Linear(784, 512), nn.ReLU(), nn.Linear(512, 256), nn.ReLU())
        self.head_mu = nn.Linear(256, latent)        # q(z|x) 의 평균 / the mean of q(z|x)
        self.head_logvar = nn.Linear(256, latent)    # q(z|x) 의 로그분산 / the log-variance of q(z|x)
        self.dec = nn.Sequential(nn.Linear(latent, 256), nn.ReLU(), nn.Linear(256, 512), nn.ReLU(), nn.Linear(512, 784))

    def encode(self, x):
        """이미지 -> q(z|x) 의 평균과 로그분산 / image -> mean and log-variance of q(z|x)"""
        h = self.enc(x)
        return self.head_mu(h), self.head_logvar(h)

    def decode(self, z):
        """코드 -> 784 개의 logit / code -> 784 logits, one per pixel"""
        return self.dec(z)


def reparameterize(mu, logvar):
    """4주차 실습 §3 에서 쓴 줄 그대로: z = mu + std * eps / the line you wrote in §3 of the Week 4 lab"""
    std = torch.exp(0.5 * logvar)
    eps = torch.randn_like(std)
    return mu + std * eps


def kl_term(mu, logvar):
    """4주차 실습 §3 의 KL( q(z|x) || p(z) ), 이미지마다 하나 / the Week 4 lab KL( q(z|x) || p(z) ), one number per image"""
    q = Normal(mu, torch.exp(0.5 * logvar))
    prior = Normal(torch.zeros_like(mu), torch.ones_like(mu))
    return kl_divergence(q, prior).sum(dim=1)


def decode_to_images(model, z):
    """logit 을 0 ~ 1 픽셀 값으로 / logits to pixel values in 0..1"""
    with torch.no_grad():
        return torch.sigmoid(model.decode(torch.as_tensor(z, dtype=torch.float32))).numpy()


def train_vae(beta_of, name):
    """4주차 실습 §4 의 반복문과 같고 한 줄만 다르다: β 가 상수가 아니라 학습 진행 t 의 함수 beta_of(t) 다.
    The Week 4 lab §4 loop with one line changed: β is a function beta_of(t) of the training progress t, not a constant.
    t 는 지금까지 지난 에폭 수(소수)다. 돌려주는 것: 모형, 에폭마다 (에폭, 복원항, KL, 에폭 끝의 β).
    t is the number of epochs so far, a fraction. Returns the model and, per epoch, (epoch, reconstruction, KL, β at the end of the epoch)."""
    torch.manual_seed(SEED)                  # 같은 시드 = 같은 처음 가중치 · 같은 자료 순서 · 같은 잡음 / same seed: same initial weights, data order and noise
    model = VAE(LATENT)
    opt = torch.optim.Adam(model.parameters(), lr=LR)
    steps_per_epoch = math.ceil(N_TRAIN / BATCH)   # 한 에폭은 235 걸음 / one epoch is 235 steps
    step, hist = 0, []
    print(f"--- {name}")
    print(f'{"epoch":>5} {"reconstruction":>15} {"KL":>8} {"beta":>6}')
    for ep in range(1, EPOCHS + 1):
        perm = torch.randperm(N_TRAIN)
        rec_tot = kl_tot = 0.0
        for i in range(0, N_TRAIN, BATCH):
            xb = X_TRAIN[perm[i:i + BATCH]]
            beta = float(beta_of(step / steps_per_epoch))   # 이 걸음의 β. 4주차와 다른 줄 / β for this step: the line that differs from Week 4
            mu, logvar = model.encode(xb)
            z = reparameterize(mu, logvar)
            rec = F.binary_cross_entropy_with_logits(model.decode(z), xb, reduction="sum") / len(xb)
            kl = kl_term(mu, logvar).mean()
            loss = rec + beta * kl
            opt.zero_grad(); loss.backward(); opt.step()
            rec_tot += rec.item() * len(xb); kl_tot += kl.item() * len(xb); step += 1
        hist.append((ep, rec_tot / N_TRAIN, kl_tot / N_TRAIN, beta))
        print(f"{ep:5d} {rec_tot / N_TRAIN:15.2f} {kl_tot / N_TRAIN:8.2f} {beta:6.3f}")
    return model, hist


def settings_now():
    """지금 쓰는 설정값 / the settings in use now"""
    return {k: globals()[k] for k in PRESCRIBED}


def run_condition(key, beta_of):
    """학습 하나를 돌리고 RUNS 에 남긴다. 오류가 나도 모두 실행을 멈추지 않는다.
    Run one training and keep it in RUNS. An error does not stop Run all."""
    t0 = time.time()
    try:
        model, hist = train_vae(beta_of, RUN_NAMES[key])
    except Exception as e:  # noqa: BLE001
        print(f"학습이 끝나지 못했습니다 / the run did not finish: {type(e).__name__}: {e}")
        RUNS.pop(key, None)
        return
    sec = time.time() - t0
    RUNS[key] = {"model": model, "hist": hist, "sec": sec, "id": f"{key}-{time.time():.3f}", "settings": settings_now()}
    print(f"걸린 시간 / time: {sec:.0f} s")
    if hist[-1][1] > 170:
        print("마지막 에폭의 복원항이 170 보다 큽니다. 학습이 제대로 되지 않았습니다. 런타임 > 모두 실행 으로 다시 돌려 보세요.")
        print("The last epoch's reconstruction is above 170, so the run did not learn. Try Runtime > Run all again.")


def evaluate(model):
    """시험 이미지 10,000장에서: 복원항(BCE)과 KL, 둘 다 이미지당 평균, 그리고 인코더 평균 mu 와 로그분산 logvar.
    On the 10,000 test images: the reconstruction (BCE) and the KL, both per image, and the encoder means mu and log-variances logvar."""
    torch.manual_seed(12345)                 # z 를 뽑는 잡음도 고정. 세 모형이 같은 잡음을 쓴다 / fix the noise that draws z; all three models use the same noise
    with torch.no_grad():
        mu, logvar = model.encode(X_TEST)
        z = reparameterize(mu, logvar)
        rec = F.binary_cross_entropy_with_logits(model.decode(z), X_TEST, reduction="sum").item() / len(X_TEST)
        kl = kl_term(mu, logvar).mean().item()
    return {"rec": rec, "kl": kl, "mu": mu, "logvar": logvar, "max_mu": mu.abs().max().item()}


print("torch", torch.__version__, "| numpy", np.__version__)
print("실행 장치 / device: cpu. GPU 가 없어도 됩니다 / no GPU needed")
```

```python
#@title ▶ 자료 / Data. MNIST 내려받기, 10초 남짓 / Download MNIST, about ten seconds { display-mode: "form" }
MNIST_FILES = ["train-images-idx3-ubyte.gz", "train-labels-idx1-ubyte.gz",
               "t10k-images-idx3-ubyte.gz", "t10k-labels-idx1-ubyte.gz"]
MIRRORS = ["https://ossci-datasets.s3.amazonaws.com/mnist/",
           "https://storage.googleapis.com/cvdf-datasets/mnist/",
           "https://raw.githubusercontent.com/fgnt/mnist/master/"]       # 4주차 실습과 같은 세 곳 / the same three mirrors as the Week 4 lab


def fetch(name):
    """먼저 응답하는 곳에서 내려받는다 / download from the first mirror that answers"""
    if os.path.exists(name):
        return name
    errors = []
    for base in MIRRORS:
        try:
            urllib.request.urlretrieve(base + name, name)
            return name
        except Exception as exc:  # noqa: BLE001
            errors.append(f"{base}: {type(exc).__name__}")
    raise RuntimeError("MNIST 내려받기 실패 / could not download " + name + "\n  " + "\n  ".join(errors))


def idx_images(path):
    with gzip.open(path, "rb") as f:
        _, n, r, c = struct.unpack(">IIII", f.read(16))
        return np.frombuffer(f.read(), np.uint8).reshape(n, r * c)


def idx_labels(path):
    with gzip.open(path, "rb") as f:
        struct.unpack(">II", f.read(8))
        return np.frombuffer(f.read(), np.uint8)


for _f in MNIST_FILES:
    fetch(_f)
X_TRAIN = torch.from_numpy(idx_images("train-images-idx3-ubyte.gz").astype(np.float32) / 255.0)   # 픽셀을 0 ~ 1 로 / pixels scaled to 0..1
X_TEST = torch.from_numpy(idx_images("t10k-images-idx3-ubyte.gz").astype(np.float32) / 255.0)
Y_TEST = idx_labels("t10k-labels-idx1-ubyte.gz")
N_TRAIN = len(X_TRAIN)
print("학습 이미지 / training images:", tuple(X_TRAIN.shape))
print("시험 이미지 / test images:    ", tuple(X_TEST.shape))
print("시험 이미지의 숫자별 개수 / test images per digit:", np.bincount(Y_TEST, minlength=10).tolist())
# 기대값 / Expected  (60000, 784), (10000, 784), [980, 1135, 1032, 1010, 982, 892, 958, 1028, 974, 1009]
```
