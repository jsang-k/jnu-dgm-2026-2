```python
#@title ▶ 준비 / Setup. 먼저 실행하세요 / Run this first { display-mode: "form" }
import math, time, json, hashlib, platform, datetime
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt

torch.set_num_threads(2)                 # 스레드 수 고정. 같은 기기에서 다시 돌리면 같은 숫자 / fix the thread count so a re-run on the same machine gives the same numbers
np.set_printoptions(precision=3, suppress=True)

NOTEBOOK = "A2 GAN track v3 (2026-10-15)"
SEED = 2                                 # 6주차 실습 §1 과 같은 시드 / the Week 6 lab §1 seed
STEPS = 10000                            # 6주차 실습과 같은 학습 길이 / the Week 6 lab's training length
R, STD = 2.0, 0.05                       # 반지름 2 인 원 위, 군집마다 표준편차 0.05 / on a circle of radius 2, standard deviation 0.05 per cluster
Z_DIM, B = 16, 256                       # 잡음 z 의 차원, 한 걸음에 쓰는 점의 수 / dimension of the noise z, points per step
RADIUS = 4 * STD                         # 군집 중심에서 0.2 안의 점은 그 군집에 든다 / a point within 0.2 of a center belongs to that cluster
MIN_COUNT = 50                           # 고정 z 1,000개 가운데 50개(5%) 이상이 든 군집을 덮었다고 센다 / a cluster with 50 or more of the 1,000 fixed-z points is covered
SNAP_STEPS = (1000, 3000, 5000, 10000)   # 고정 z 로 점을 찍어 두는 걸음 / steps at which the fixed-z points are kept
PROBE_STEPS = (1, 2, 5, 10, 20, 50) + tuple(range(100, STEPS + 1, 100))   # D 의 판정을 기록하는 걸음 / steps at which D's verdicts are recorded
PRESCRIBED = {"SEED": 2, "STEPS": 10000, "R": 2.0, "STD": 0.05, "Z_DIM": 16, "B": 256, "RADIUS": 0.2, "MIN_COUNT": 50}
RUN_NAMES = {"ns": "run 1 · non-saturating (baseline)", "mm": "run 2 · minimax"}
HEAD = {"ns": "1 · non-saturating", "mm": "2 · minimax"}   # 표와 그림의 머리 / column heads for tables and figures
COLORS = {"ns": "#406551", "mm": "#B34F3F"}               # 강의 슬라이드의 초록과 빨강 / the slides' green and red
GREEN, GOLD = "#406551", "#C9971C"
RESULTS, RUNS = {}, {}                   # 이번 실행의 결과와 학습 기록 / this run's results and training records

CENTERS = torch.tensor([[R * math.cos(2 * math.pi * k / 8), R * math.sin(2 * math.pi * k / 8)]
                        for k in range(8)])   # 군집 중심 8개. 크기 (8, 2) / the 8 cluster centers, shape (8, 2)
Z_FIX = torch.randn(1000, Z_DIM, generator=torch.Generator().manual_seed(1234))   # 그림마다 같은 z 1,000개 / the same 1,000 z for every picture


def sample_ring(n):
    """진짜 점 n 개. 군집을 하나 고르고 그 중심 둘레에 작은 잡음을 더한다.
    n real points. Pick a cluster, then add small noise around its center."""
    k = torch.randint(0, 8, (n,))
    return CENTERS[k] + STD * torch.randn(n, 2)


def make_G():
    """생성자 G. 잡음 z(숫자 16개)를 점 하나(숫자 2개)로 / Generator G: noise z (16 numbers) to one point (2 numbers)"""
    return nn.Sequential(nn.Linear(Z_DIM, 128), nn.ReLU(), nn.Linear(128, 128), nn.ReLU(), nn.Linear(128, 2))


def make_D():
    """판별자 D. 점 하나에 로짓 하나. 시그모이드는 bce 안에 있다 / Discriminator D: one logit per point; the sigmoid lives inside bce"""
    return nn.Sequential(nn.Linear(2, 128), nn.ReLU(), nn.Linear(128, 128), nn.ReLU(), nn.Linear(128, 1))


def bce(logits, target):
    """이진 교차엔트로피. 로짓에서 바로 계산한다. bce(로짓, 1) = 평균 −log σ(a), bce(로짓, 0) = 평균 −log(1 − σ(a)).
    Binary cross-entropy computed straight from the logits: bce(logit, 1) = mean −log σ(a), bce(logit, 0) = mean −log(1 − σ(a))."""
    return F.binary_cross_entropy_with_logits(logits, torch.full_like(logits, float(target)))


def ref_d_loss(D, x_real, x_fake):
    """단계 A. D 의 손실. 6주차 실습 §2 의 답. 두 학습에서 같다 / step A, D's loss: the Week 6 lab §2 answer, the same in both runs"""
    return bce(D(x_real), 1) + bce(D(x_fake.detach()), 0)


def g_loss_ns(D, x_fake):
    """단계 B. 비포화 G 손실, 학습 1. 6주차 실습 §2 의 답 / step B, the non-saturating G loss of run 1: the Week 6 lab §2 answer"""
    return bce(D(x_fake), 1)


def train_gan(g_loss_fn, name):
    """6주차 실습의 학습 그대로. 한 걸음 = D 한 번 갱신(단계 A), 그다음 같은 x_fake 로 G 한 번 갱신(단계 B).
    The Week 6 lab training: one step = one D update (step A), then one G update on the same x_fake (step B).
    두 학습 사이에 다른 것은 G 의 손실 g_loss_fn 하나 / the only difference between the two runs is G's loss, g_loss_fn.
    돌려주는 것: 고정 z 로 찍은 점, D 의 판정(로짓) 기록, 100 걸음마다의 두 손실.
    Returns the fixed-z points, the record of D's verdicts (logits) and the two losses every 100 steps."""
    torch.manual_seed(SEED)                  # 같은 시드 = 같은 처음 가중치, 같은 진짜 점, 같은 잡음 / same seed: same initial weights, real points and noise
    G, D = make_G(), make_D()
    opt_G = torch.optim.Adam(G.parameters(), lr=1e-3, betas=(0.5, 0.999))   # G 만 움직인다 / moves only G
    opt_D = torch.optim.Adam(D.parameters(), lr=1e-3, betas=(0.5, 0.999))   # D 만 움직인다 / moves only D
    probe_at = set(PROBE_STEPS)
    snaps, probes, hist = {}, {}, []
    print(f"--- {name}")
    for step in range(1, STEPS + 1):
        x_real = sample_ring(B)              # 진짜 점 256개 / 256 real points
        z = torch.randn(B, Z_DIM)            # 잡음 256개 / 256 noise vectors
        x_fake = G(z)                        # 생성한 점 256개 / 256 generated points
        d_loss = ref_d_loss(D, x_real, x_fake)   # 단계 A. D 의 손실 / step A, D's loss
        opt_D.zero_grad(); d_loss.backward(); opt_D.step()
        if step in probe_at:
            with torch.no_grad():
                probes[step] = D(x_fake).squeeze(1)   # 단계 B 의 G 가 받을 D 의 판정(로짓) / the verdicts (logits) of the D that step B uses
        g_loss = g_loss_fn(D, x_fake)        # 단계 B. 같은 x_fake 로 G 의 손실. 두 학습에서 다른 줄 / step B, G's loss on the same x_fake: the line that differs
        opt_G.zero_grad(); g_loss.backward(); opt_G.step()
        if step % 100 == 0:
            hist.append((step, d_loss.item(), g_loss.item()))
        if step in SNAP_STEPS:
            with torch.no_grad():
                snaps[step] = G(Z_FIX)       # 고정 z 1,000개로 만든 점 / the points made from the 1,000 fixed z
            print(f"  step {step:6,d}   D loss {d_loss.item():.3f}   G loss {g_loss.item():.3f}")
    return snaps, probes, hist


def settings_now():
    """지금 쓰는 설정값 / the settings in use now"""
    return {k: globals()[k] for k in PRESCRIBED}


def run_condition(key, g_loss_fn):
    """학습 하나를 돌리고 RUNS 에 남긴다. 오류가 나도 모두 실행을 멈추지 않는다.
    Run one training and keep it in RUNS. An error does not stop Run all."""
    t0 = time.time()
    try:
        snaps, probes, hist = train_gan(g_loss_fn, RUN_NAMES[key])
    except Exception as e:  # noqa: BLE001
        print(f"학습이 끝나지 못했습니다 / the run did not finish: {type(e).__name__}: {e}")
        RUNS.pop(key, None)
        return
    sec = time.time() - t0
    RUNS[key] = {"snaps": snaps, "probes": probes, "hist": hist, "sec": sec, "id": f"{key}-{time.time():.3f}", "settings": settings_now()}
    print(f"걸린 시간 / time: {sec:.0f} s")
    if not all(torch.isfinite(x).all() for x in snaps.values()):
        print("생성한 점에 nan 이나 inf 가 있습니다. 학습이 무너졌습니다. 코드 1 을 다시 보세요.")
        print("The generated points contain nan or inf, so training broke down. Look at Code 1 again.")


print("torch", torch.__version__, "| numpy", np.__version__)
print("실행 장치 / device: cpu. GPU 가 없어도 됩니다 / no GPU needed")
```

```python
#@title §3 ▶ D 의 판정 기록 / The record of D's verdicts { display-mode: "form" }
if not RUNS:
    print("아직 학습이 없습니다. / No training run yet.")
else:
    cols = [k for k in RUN_NAMES if k in RUNS]
    own = {"ns": "push_ns", "mm": "push_minimax"}
    other = {"ns": "push_minimax", "mm": "push_ns"}
    PROBE_SUM = {}
    for k in cols:
        st = sorted(RUNS[k]["probes"])
        p = {s: torch.sigmoid(RUNS[k]["probes"][s]) for s in st}
        PROBE_SUM[k] = {"steps": st, "p": [p[s].mean().item() for s in st], "sure": [(p[s] < 0.1).float().mean().item() for s in st]}
        if code_usable("Code2"):
            for name, fn in (("own", own[k]), ("other", other[k])):
                vals = []
                for s in st:
                    out, err = _call(fn, RUNS[k]["probes"][s])
                    vals.append(torch.as_tensor(out, dtype=torch.float32).mean().item() if (err is None and out is not None) else float("nan"))
                PROBE_SUM[k][name] = vals
    fig, axes = plt.subplots(1, 3, figsize=(15, 4.0))
    for k in cols:
        st = PROBE_SUM[k]["steps"]
        axes[0].plot(st, PROBE_SUM[k]["p"], "-", color=COLORS[k], label=HEAD[k])
        axes[1].plot(st, PROBE_SUM[k]["sure"], "-", color=COLORS[k], label=HEAD[k])
        if "own" in PROBE_SUM[k]:
            axes[2].plot(st, PROBE_SUM[k]["own"], "-", color=COLORS[k], label=f"{HEAD[k]}: its own loss")
            axes[2].plot(st, PROBE_SUM[k]["other"], ":", color=COLORS[k], label=f"{HEAD[k]}: the other loss, same points")
    axes[0].set_title("mean D(G(z)) over the 256 generated points"); axes[0].set_ylim(0, 1)
    axes[1].set_title("share of generated points with D(G(z)) < 0.1"); axes[1].set_ylim(0, 1)
    axes[2].set_title("mean push on the logit, |dL/da|" if code_usable("Code2") else "push: after Code 2"); axes[2].set_ylim(0, 1)
    for a in axes:
        a.set_xscale("log"); a.set_xlabel("step (log scale)"); a.grid(alpha=0.3)
        if a.get_legend_handles_labels()[0]:
            a.legend(fontsize=8, loc="lower left" if a is axes[2] else "upper right")
    plt.tight_layout(); plt.show()
    show = [1, 10, 100, 1000, 3000, 10000]
    print("가로축은 로그 눈금입니다. 처음 100 걸음이 그림의 절반을 차지합니다. / The x-axis is a log scale; the first 100 steps take up half of each panel.")
    print("\n" + _pad("걸음 / step", 34) + "".join(f"{s:>9,d}" for s in show))
    for k in cols:
        st = PROBE_SUM[k]["steps"]
        pick = lambda name: "".join(f"{PROBE_SUM[k][name][st.index(s)]:9.3f}" for s in show)
        print(_pad(f"{HEAD[k]}  D(G(z))", 34) + pick("p"))
        print(_pad(f"{HEAD[k]}  D(G(z)) < 0.1", 34) + pick("sure"))
        if "own" in PROBE_SUM[k]:
            print(_pad(f"{HEAD[k]}  push, own loss", 34) + pick("own"))
    print("\nD(G(z)) 는 D 가 생성한 점을 진짜라고 볼 확률입니다. 0 에 가까우면 D 가 그 점을 확실히 가짜로 보고, 0.5 근처면 진짜와 가짜의 중간으로 봅니다.")
    print("D(G(z)) is the probability D gives that a generated point is real: near 0, D is sure the point is generated; near 0.5, D rates it halfway between real and generated.")
    if code_usable("Code2"):
        print("minimax 손실이 로짓을 미는 크기는 σ(a) = D(G(z)) 이므로, 학습 2 의 push 줄은 D(G(z)) 줄과 같습니다.")
        print("The minimax push is σ(a) = D(G(z)), so run 2's push row equals its D(G(z)) row.")
    if not code_usable("Code2"):
        print("코드 2 가 통과하거나 기준 코드로 바뀌면 오른쪽 그림과 push 줄이 나옵니다. / The right panel and the push rows appear once Code 2 passes or is switched to the reference code.")
    record("Probe", ids={k: RUNS[k]["id"] for k in cols},
           probe={k: {name: [round(v, 4) for v in PROBE_SUM[k][name]] for name in ("p", "sure", "own", "other") if name in PROBE_SUM[k]} for k in cols},
           steps=PROBE_SUM[cols[0]]["steps"])
```

```python
#@title ▶ 분석 영수증 / Analysis receipt { display-mode: "form" }
ai_use = ""  #@param {type:"string"}

pred = RESULTS.get("Pred", {})
cover = RESULTS.get("Cover", {})
probe = RESULTS.get("Probe", {})
now_pred = {k: (globals().get(k) or "").strip() if isinstance(globals().get(k), str) else "" for k in PRED_FIELDS}
why = []
if pred and any(pred.get(k) != v for k, v in now_pred.items()):
    why.append(("예측 칸이 바뀐 뒤 그 셀을 다시 실행하지 않았습니다", "the prediction cell was edited but not run again"))
changed = [k for k in PRESCRIBED if settings_now()[k] != PRESCRIBED[k]]
if changed or any(RUNS[k]["settings"] != PRESCRIBED for k in RUNS):
    why.append(("정해진 설정이 바뀌었습니다: " + ", ".join(changed or ["학습 때 설정"]), "the prescribed settings changed: " + ", ".join(changed or ["settings at training time"])))
stale = [k for k in RUNS if (cover and cover.get("ids", {}).get(k) != RUNS[k]["id"]) or (probe and probe.get("ids", {}).get(k) != RUNS[k]["id"])]
late = (RUNS and ((code_usable("Code3") and not cover) or not probe
                  or (code_usable("Code2") and any("own" not in v for v in probe.get("probe", {}).values()))))
if stale or late:
    why.append(("덮은 군집 표나 판정 기록이 지금의 학습과 코드 칸으로 만든 것이 아닙니다", "the coverage table or the verdict record was not built from the current runs and code cells"))
consistent = not why

todo = []
if not predictions_ready() or not pred:
    todo.append(("Q1(b) 학번과 예측 채우기", "fill in your student ID and predictions in Q1(b)"))
missing = [k for k in RUN_NAMES if k not in RUNS]
if missing:
    todo.append(("학습 실행하기: " + ", ".join(RUN_NAMES[k].split(" · ")[0] for k in missing), "run: " + ", ".join(RUN_NAMES[k].split(" · ")[0] for k in missing)))
left = [c for c in CODE_CELLS if not code_usable(c)]
refd = [c for c in CODE_CELLS if code_usable(c) and not code_ok(c)]
if left:
    todo.append(("코드 칸 통과시키기(또는 기준 코드로 바꾸기): " + ", ".join(f"코드 {c[-1]}" for c in left), "make the code cells pass (or switch them to the reference code): " + ", ".join(f"Code {c[-1]}" for c in left)))
if not ai_use.strip():
    todo.append(("AI 사용 칸 채우기(안 썼으면 none)", "fill in AI use (none if you used no AI)"))
ready = consistent and not todo

body, hashed = [], []


def add(text, plain=None):
    """영수증 한 줄. 지문은 학습 시간을 뺀 plain 으로 낸다 / one receipt line; the fingerprint uses plain, without the training time"""
    body.append(text)
    hashed.append(text if plain is None else plain)


add("일관성 / consistency         " + (("OK. 모든 결과가 지금의 예측과 설정과 학습에서 나옴 / everything comes from the current predictions, settings and runs" if RUNS
                                       else "OK. 아직 학습이 없음 / no runs yet") if consistent
     else "NOT OK. " + "; ".join(f"{k} / {e}" for k, e in why) + ". 런타임 > 모두 실행 / Runtime > Run all"))
add("제출 준비 / submission       " + ("READY" if ready else "NOT READY. 남은 일 / still to do: " + "; ".join(f"{k} / {e}" for k, e in todo)))
add(f"노트북 / notebook            {NOTEBOOK}")
add(f"학번 / student ID            {now_pred['STUDENT_ID'] or '(비어 있음 / empty)'}")
add(f"예측 / predictions           결과 전에 씀 / written before running: {'yes' if pred.get('before') else 'no'}")
for k, label in [("PREDICT_EARLY", "1 early"), ("PREDICT_END", "2 at the end"), ("MECHANISM", "3 mechanism"), ("DIAGNOSTIC", "4 diagnostic")]:
    add(f"    {label:15s}          {pred.get(k) or '(비어 있음 / empty)'}")
s = settings_now()
add(f"설정 / settings              seed {s['SEED']} · steps {s['STEPS']} · ring R {s['R']} STD {s['STD']} · z {s['Z_DIM']} · batch {s['B']} · radius {s['RADIUS']} · min count {s['MIN_COUNT']}")
add("코드 칸 / code cells          " + " · ".join(f"{c[-1]} " + ("OK" if code_ok(c) else ("기준 코드 / reference" if code_usable(c) else "미완 / not done")) for c in CODE_CELLS))
add("학습 / training               마지막 기록, 걸음 10,000 / last record, step 10,000")
for k in RUN_NAMES:
    if k in RUNS:
        st, dl, gl = RUNS[k]["hist"][-1]
        add(f"    {HEAD[k]:20s}     D loss {dl:.3f} · G loss {gl:.3f} · {RUNS[k]['sec']:.0f} s",
            f"    {HEAD[k]:20s}     D loss {dl:.3f} · G loss {gl:.3f}")
    else:
        add(f"    {HEAD[k]:20s}     실행 안 됨 / not run")
cv = cover.get("cover", {})
if cv:
    cols = [k for k in RUN_NAMES if k in cv]
    add("덮은 군집 / coverage         " + " | ".join(f"{HEAD[k]}: covered, off-cluster" for k in cols))
    for st in SNAP_STEPS:
        add(f"    step {st:6,d}              " + " | ".join(f"{cv[k][st]['covered']} of 8, {cv[k][st]['off']:.1%}" for k in cols))
    for k in cols:
        add(f"    {HEAD[k]:20s} 10,000 걸음 군집별 / per cluster at 10,000: {cv[k][STEPS]['counts']}")
else:
    add("덮은 군집 / coverage         아직 없음 / not yet")
pr = probe.get("probe", {})
if pr:
    stp = probe["steps"]
    show = [1, 10, 100, 1000, 10000]
    add("판정 기록 / verdicts         걸음 / steps " + ", ".join(f"{v:,}" for v in show))
    for k in [k for k in RUN_NAMES if k in pr]:
        add(f"    {HEAD[k]:20s} D(G(z))       " + "  ".join(f"{pr[k]['p'][stp.index(v)]:.3f}" for v in show)
            + f"   (가장 낮은 값 / lowest {min(pr[k]['p']):.3f})")
        add(f"    {HEAD[k]:20s} D(G(z)) < 0.1 " + "  ".join(f"{pr[k]['sure'][stp.index(v)]:.3f}" for v in show)
            + f"   (가장 높은 값 / highest {max(pr[k]['sure']):.3f})")
        if "own" in pr[k]:
            early = [v for v, t in zip(pr[k]["own"], stp) if t <= 1000]
            add(f"    {HEAD[k]:20s} push, own     " + "  ".join(f"{pr[k]['own'][stp.index(v)]:.3f}" for v in show)
                + f"   (처음 1,000 걸음 기록의 평균 / mean of the records in the first 1,000 steps {sum(early) / len(early):.3f})")
else:
    add("판정 기록 / verdicts         아직 없음 / not yet")
try:
    import subprocess
    gpu = subprocess.run(["nvidia-smi", "-L"], capture_output=True, text=True, timeout=5).stdout.strip()
except Exception:  # noqa: BLE001
    gpu = ""
add(f"환경 / environment           Python {platform.python_version()} · torch {torch.__version__} · numpy {np.__version__} · "
    + ("GPU runtime, the model ran on CPU" if gpu else "CPU"))
add(f"AI 사용 / AI use              {ai_use.strip() if ai_use.strip() else '(비어 있음 / empty)'}")

fingerprint = hashlib.sha256("\n".join(hashed).encode("utf-8")).hexdigest()[:10]
bar = "=" * 96
print(bar)
print("  분석 영수증 / Analysis receipt.  이 칸 전체를 PDF 에 글자 그대로 붙이세요 / Paste this whole block into your PDF as text")
print(bar)
for line in body:
    print(line)
print(f"실행 시각 / run at            {datetime.datetime.now().strftime('%Y-%m-%d %H:%M')}")
print(f"지문 / fingerprint           {fingerprint}")
print(bar)
if ready:
    print("READY 는 예측, 두 학습, 코드 칸 셋(통과했거나 기준 코드로 바꾼 칸), AI 사용 칸이 모두 갖춰졌다는 뜻입니다. 서술형 답과 이 영수증을 PDF 에 넣어야 제출이 끝납니다.")
    print("READY means your predictions, the two runs, the three code cells (passed, or switched to the reference code) and the AI-use line are all in place. You still need your written answers and this receipt in your PDF.")
    if refd:
        print("기준 코드를 쓴 칸: " + ", ".join(f"코드 {c[-1]}" for c in refd) + ". 감점은 없고, PDF 에서 그 칸의 계산도 똑같이 설명합니다.")
        print("Cells on the reference code: " + ", ".join(f"Code {c[-1]}" for c in refd) + ". There is no penalty; explain their calculation in your PDF as for the other cells.")
else:
    print("아직 NOT READY 입니다. 위 「제출 준비」 줄의 남은 일을 끝내고 런타임 > 모두 실행 을 하세요.")
    print("Still NOT READY. Finish what the submission line lists, then Runtime > Run all.")
```
